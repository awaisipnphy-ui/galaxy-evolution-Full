```python
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

# -----------------------------
# Plot style
# -----------------------------
plt.rcParams.update({
    "figure.figsize": (7, 5),
    "axes.labelsize": 13,
    "axes.titlesize": 13,
    "legend.fontsize": 10,
    "xtick.labelsize": 11,
    "ytick.labelsize": 11,
    "lines.linewidth": 2.0,
    "savefig.bbox": "tight",
    "axes.grid": False,
})

outdir = Path("paper_figures")
outdir.mkdir(exist_ok=True)
print("Saving figures in:", outdir.resolve())

# -----------------------------
# Numerical parameters
# -----------------------------
a = 1.0
N = 60
ell_plot = 2
M = 10.0

m_scalar = 0.0
xi = 0.0

# Number of outermost points to remove from Fig. 1c only
trim = 2

# -----------------------------
# Metric functions
# -----------------------------
def flat_metric_f(r):
    return np.ones_like(r)

def flat_metric_g(r):
    return np.ones_like(r)

def schwarzschild_f(r, M):
    return 1.0 - 2.0 * M / r

def schwarzschild_g(r, M):
    return 1.0 - 2.0 * M / r

def schwarzschild_iso_C(rho, M):
    return (1.0 - M / (2.0 * rho)) / (1.0 + M / (2.0 * rho))

def schwarzschild_iso_B(rho, M):
    return (1.0 + M / (2.0 * rho))**2

def schwarzschild_iso_areal_radius(rho, M):
    return rho * (1.0 + M / (2.0 * rho))**2

# -----------------------------
# Coefficient builders
# -----------------------------
def build_static_FGU(r, ell, f, g):
    """
    Static-formulation coefficients:
        F = r^2 sqrt(f g)
        G = (f g)^(1/4) / r
        U = ell(ell+1) f / r^2
    """
    ff = f(r)
    gg = g(r)

    if np.any(ff <= 0) or np.any(gg <= 0):
        raise ValueError("Metric functions must stay positive on the chosen lattice.")

    F = r**2 * np.sqrt(ff * gg)
    G = (ff * gg)**0.25 / r
    U = ell * (ell + 1.0) * ff / r**2
    return F, G, U

def build_isotropic_lattice_coeffs(rho, ell, Cfunc, Bfunc, M, m_scalar=0.0, xi=0.0):
    """
    Isotropic-formulation lattice coefficients from the manuscript:
        F_iso = C B
        W_iso = sqrt(C / B^3)
        V_iso = C^2/B^3 * (CB)'/rho + C^2/B^2 * ell(ell+1)/rho^2 + C^2 (m^2 + xi R)

    For vacuum isotropic Schwarzschild outside the source:
        R = 0
    """
    C = Cfunc(rho, M)
    B = Bfunc(rho, M)
    R = np.zeros_like(rho)

    if np.any(C <= 0) or np.any(B <= 0):
        raise ValueError("C and B must stay positive on the chosen isotropic lattice.")

    CB = C * B
    dCB = np.gradient(CB, rho, edge_order=2)

    F_iso = C * B
    W_iso = np.sqrt(C / B**3)
    V_iso = (
        (C**2 / B**3) * (dCB / rho)
        + (C**2 / B**2) * ell * (ell + 1.0) / rho**2
        + C**2 * (m_scalar**2 + xi * R)
    )
    return F_iso, W_iso, V_iso

# -----------------------------
# Lattice matrix builder
# -----------------------------
def build_K_from_coeffs(F, G, U, a):
    """
    Build the tridiagonal lattice stiffness matrix K from
    the coefficient triplet (F, G, U).
    """
    Nloc = len(U)
    K = np.zeros((Nloc, Nloc), dtype=float)

    # off-diagonal
    for j in range(Nloc - 1):
        off = -F[j] * G[j] * G[j + 1] / a**2
        K[j, j + 1] = off
        K[j + 1, j] = off

    # diagonal
    if Nloc >= 2:
        K[0, 0] = U[0] + F[0] * G[0]**2 / a**2
        K[-1, -1] = U[-1] + F[-2] * G[-1]**2 / a**2
        for j in range(1, Nloc - 1):
            K[j, j] = U[j] + (F[j] + F[j - 1]) * G[j]**2 / a**2
    else:
        K[0, 0] = U[0]

    K = 0.5 * (K + K.T)

    # small stabilization if numerical roundoff creates tiny negative eigenvalues
    evals = np.linalg.eigvalsh(K)
    if evals.min() <= 0:
        K += np.eye(Nloc) * (1e-12 - evals.min() + 1e-12)

    return K

# -----------------------------
# Helper for log plots
# -----------------------------
def positive_clip(x, eps=1e-14):
    return np.clip(x, eps, None)

# -----------------------------
# Radial lattices
# -----------------------------
r_flat = a * np.arange(1, N + 1)

# start safely outside horizon r = 2M
r_curv = (2.5 * M) + a * np.arange(N)

# start safely outside isotropic horizon rho = M/2
rho_iso = (0.6 * M) + a * np.arange(N)

r_areal_iso = schwarzschild_iso_areal_radius(rho_iso, M)

print("Flat lattice range:", r_flat[0], "to", r_flat[-1])
print("Curvature Schwarzschild range:", r_curv[0], "to", r_curv[-1])
print("Isotropic Schwarzschild range:", rho_iso[0], "to", rho_iso[-1])
print("Isotropic areal-radius range:", r_areal_iso[0], "to", r_areal_iso[-1])

# -----------------------------
# Compute coefficients
# -----------------------------
# Static formulation
F_flat, G_flat, U_flat = build_static_FGU(
    r_flat, ell_plot,
    flat_metric_f, flat_metric_g
)

F_curv, G_curv, U_curv = build_static_FGU(
    r_curv, ell_plot,
    lambda r: schwarzschild_f(r, M),
    lambda r: schwarzschild_g(r, M)
)

# Isotropic formulation
F_iso, W_iso, V_iso = build_isotropic_lattice_coeffs(
    rho_iso, ell_plot,
    schwarzschild_iso_C, schwarzschild_iso_B, M,
    m_scalar=m_scalar, xi=xi
)

# -----------------------------
# Figure 1a: static coefficients only
# -----------------------------
fig, axes = plt.subplots(3, 1, figsize=(7, 10), sharex=False)

axes[0].plot(r_flat, positive_clip(F_flat), label="Flat")
axes[0].plot(r_curv, positive_clip(F_curv), label="Schwarzschild (curvature)")
axes[0].set_ylabel(r"$F(r)$")
axes[0].set_yscale("log")
axes[0].legend()

axes[1].plot(r_flat, positive_clip(G_flat), label="Flat")
axes[1].plot(r_curv, positive_clip(G_curv), label="Schwarzschild (curvature)")
axes[1].set_ylabel(r"$G(r)$")
axes[1].set_yscale("log")

axes[2].plot(r_flat, positive_clip(U_flat), label="Flat")
axes[2].plot(r_curv, positive_clip(U_curv), label="Schwarzschild (curvature)")
axes[2].set_ylabel(r"$U_\ell(r)$")
axes[2].set_xlabel(r"areal radius $r$")
axes[2].set_yscale("log")

fig.tight_layout()
fig.savefig(outdir / "figure1a_static_coefficients.pdf")
fig.savefig(outdir / "figure1a_static_coefficients.png", dpi=300)
plt.show()

# -----------------------------
# Figure 1b: isotropic coefficients separately
# -----------------------------
fig, axes = plt.subplots(3, 1, figsize=(7, 10), sharex=True)

axes[0].plot(rho_iso, positive_clip(F_iso))
axes[0].set_ylabel(r"$F_{\rm iso}(\rho)=C(\rho)B(\rho)$")
axes[0].set_yscale("log")

axes[1].plot(rho_iso, positive_clip(W_iso))
axes[1].set_ylabel(r"$W_{\rm iso}(\rho)=\sqrt{C/B^3}$")
axes[1].set_yscale("log")

axes[2].plot(rho_iso, positive_clip(V_iso))
axes[2].set_ylabel(r"$V_{\ell,{\rm iso}}(\rho)$")
axes[2].set_xlabel(r"isotropic radius $\rho$")
axes[2].set_yscale("log")

fig.tight_layout()
fig.savefig(outdir / "figure1b_isotropic_coefficients.pdf")
fig.savefig(outdir / "figure1b_isotropic_coefficients.png", dpi=300)
plt.show()

# -----------------------------
# Figure 1c: physically meaningful comparison through K entries
# -----------------------------
K_curv = build_K_from_coeffs(F_curv, G_curv, U_curv, a)
K_iso  = build_K_from_coeffs(F_iso, W_iso, V_iso, a)

diag_curv = np.diag(K_curv)
diag_iso  = np.diag(K_iso)

off_curv = -np.diag(K_curv, 1)
off_iso  = -np.diag(K_iso, 1)

# site and bond coordinates
r_site_curv = r_curv
r_site_iso  = r_areal_iso

r_bond_curv = 0.5 * (r_curv[:-1] + r_curv[1:])
r_bond_iso  = 0.5 * (r_areal_iso[:-1] + r_areal_iso[1:])

# -----------------------------
# Apply trimming only to Figure 1c
# -----------------------------
if trim > 0:
    r_site_curv_plot = r_site_curv[:-trim]
    r_site_iso_plot  = r_site_iso[:-trim]
    diag_curv_plot   = diag_curv[:-trim]
    diag_iso_plot    = diag_iso[:-trim]

    r_bond_curv_plot = r_bond_curv[:-trim]
    r_bond_iso_plot  = r_bond_iso[:-trim]
    off_curv_plot    = off_curv[:-trim]
    off_iso_plot     = off_iso[:-trim]
else:
    r_site_curv_plot = r_site_curv
    r_site_iso_plot  = r_site_iso
    diag_curv_plot   = diag_curv
    diag_iso_plot    = diag_iso

    r_bond_curv_plot = r_bond_curv
    r_bond_iso_plot  = r_bond_iso
    off_curv_plot    = off_curv
    off_iso_plot     = off_iso

# overlap region in areal radius for a fair visual comparison
common_min = max(r_site_curv_plot.min(), r_site_iso_plot.min())
common_max = min(r_site_curv_plot.max(), r_site_iso_plot.max())

fig, axes = plt.subplots(2, 1, figsize=(7, 8), sharex=False)

axes[0].plot(r_site_curv_plot, positive_clip(diag_curv_plot), label="Schwarzschild (curvature)")
axes[0].plot(r_site_iso_plot,  positive_clip(diag_iso_plot),  label="Schwarzschild (isotropic)")
axes[0].set_ylabel(r"$K_{jj}$")
axes[0].set_yscale("log")
axes[0].set_xlim(common_min, common_max)
axes[0].legend()

axes[1].plot(r_bond_curv_plot, positive_clip(off_curv_plot), label="Schwarzschild (curvature)")
axes[1].plot(r_bond_iso_plot,  positive_clip(off_iso_plot),  label="Schwarzschild (isotropic)")
axes[1].set_ylabel(r"$-K_{j,j+1}$")
axes[1].set_xlabel(r"areal radius")
axes[1].set_yscale("log")
axes[1].set_xlim(common_min, common_max)

fig.tight_layout()
fig.savefig(outdir / "figure1c_K_entries_areal_comparison_trimmed.pdf")
fig.savefig(outdir / "figure1c_K_entries_areal_comparison_trimmed.png", dpi=300)
plt.show()

print("Saved:")
print(" - figure1a_static_coefficients.(pdf/png)")
print(" - figure1b_isotropic_coefficients.(pdf/png)")
print(" - figure1c_K_entries_areal_comparison_trimmed.(pdf/png)")
```

```python
import time
import numpy as np
import matplotlib.pyplot as plt
from scipy.linalg import eigh
from pathlib import Path

# =========================================================
# Plot style
# =========================================================
plt.rcParams.update({
    "figure.figsize": (7, 5),
    "axes.labelsize": 13,
    "axes.titlesize": 13,
    "legend.fontsize": 10,
    "xtick.labelsize": 11,
    "ytick.labelsize": 11,
    "lines.linewidth": 2.0,
    "savefig.bbox": "tight",
    "axes.grid": False,
})

outdir = Path("paper_figures")
outdir.mkdir(exist_ok=True)
print("Saving figures in:", outdir.resolve())

t0 = time.time()

# =========================================================
# Numerical parameters
# =========================================================
a = 1.0
N = 120
r_flat = a * np.arange(1, N + 1)

# Safe subsystem range
n_values = np.arange(5, 31, 2)   # 5, 7, ..., 29
n_tail = 15

# Final angular scan
lmax_values = [40, 80, 120, 160, 200, 240, 280]
lmax_ref = max(lmax_values)

print("a =", a)
print("N =", N)
print("n_values =", n_values)
print("n_tail =", n_tail)
print("lmax_values =", lmax_values)

# =========================================================
# Flat-space metric and coefficient builders
# =========================================================
def flat_metric_f(r):
    return np.ones_like(r)

def flat_metric_g(r):
    return np.ones_like(r)

def build_static_FGU(r, ell, f, g):
    ff = f(r)
    gg = g(r)

    if np.any(ff <= 0) or np.any(gg <= 0):
        raise ValueError("Metric functions must stay positive on the chosen lattice.")

    F = r**2 * np.sqrt(ff * gg)
    G = (ff * gg)**0.25 / r
    U = ell * (ell + 1.0) * ff / r**2
    return F, G, U

def build_K_from_coeffs(F, G, U, a):
    Nloc = len(U)
    K = np.zeros((Nloc, Nloc), dtype=float)

    # Off-diagonal terms
    for j in range(Nloc - 1):
        off = -F[j] * G[j] * G[j + 1] / a**2
        K[j, j + 1] = off
        K[j + 1, j] = off

    # Diagonal terms
    if Nloc >= 2:
        K[0, 0] = U[0] + F[0] * G[0]**2 / a**2
        K[-1, -1] = U[-1] + F[-2] * G[-1]**2 / a**2
        for j in range(1, Nloc - 1):
            K[j, j] = U[j] + (F[j] + F[j - 1]) * G[j]**2 / a**2
    else:
        K[0, 0] = U[0]

    K = 0.5 * (K + K.T)

    evals = np.linalg.eigvalsh(K)
    if evals.min() <= 0:
        K += np.eye(Nloc) * (1e-12 - evals.min() + 1e-12)

    return K

# =========================================================
# Ground-state covariances and subsystem entropy
# =========================================================
def ground_state_covariances(K):
    w, v = eigh(K)
    w = np.clip(w, 1e-15, None)
    sqrtK = (v * np.sqrt(w)) @ v.T
    invsqrtK = (v * (1.0 / np.sqrt(w))) @ v.T
    X = 0.5 * invsqrtK
    P = 0.5 * sqrtK
    return X, P

def subsystem_entropy_from_covariances(X, P, n_sub):
    XA = X[:n_sub, :n_sub]
    PA = P[:n_sub, :n_sub]

    eigvals = np.linalg.eigvals(XA @ PA)
    eigvals = np.real_if_close(eigvals, tol=1000)
    eigvals = np.real(eigvals)
    eigvals = np.clip(eigvals, 0.25, None)

    nus = np.sqrt(eigvals)
    nus = np.clip(nus, 0.5 + 1e-14, None)

    S = np.sum(
        (nus + 0.5) * np.log(nus + 0.5)
        - (nus - 0.5) * np.log(nus - 0.5)
    )
    return float(np.real(S))

# =========================================================
# Build and cache K^(ell), X^(ell), P^(ell)
# =========================================================
def flat_K(ell):
    F, G, U = build_static_FGU(r_flat, ell, flat_metric_f, flat_metric_g)
    return build_K_from_coeffs(F, G, U, a)

K_cache = {}
X_cache = {}
P_cache = {}

t_build = time.time()
for ell in range(lmax_ref + 1):
    K = flat_K(ell)
    X, P = ground_state_covariances(K)
    K_cache[ell] = K
    X_cache[ell] = X
    P_cache[ell] = P
print(f"Cached K, X, P up to ell = {lmax_ref} in {time.time() - t_build:.2f} s")

# =========================================================
# Cache S_ell(n) for all ell and subsystem sizes
# =========================================================
S_ell_cache = {}
t_entropy = time.time()
for ell in range(lmax_ref + 1):
    X = X_cache[ell]
    P = P_cache[ell]
    S_ell_cache[ell] = np.array(
        [subsystem_entropy_from_covariances(X, P, int(n)) for n in n_values],
        dtype=float
    )
print(f"Cached S_ell(n) for ell = 0 ... {lmax_ref} in {time.time() - t_entropy:.2f} s")

# =========================================================
# Build cumulative entropy curves
# =========================================================
S_curves = {}
running = np.zeros_like(n_values, dtype=float)

for ell in range(lmax_ref + 1):
    running = running + (2 * ell + 1) * S_ell_cache[ell]
    if ell in lmax_values:
        S_curves[ell] = running.copy()

print("Stored entropy curves for:", list(S_curves.keys()))

# =========================================================
# Error diagnostics
# =========================================================
def relative_error_percent(S_test, S_ref):
    denom = 0.5 * (S_test + S_ref)
    rel = np.abs(S_test - S_ref) / np.where(denom > 0, denom, 1.0)
    return 100.0 * np.max(rel)

# Successive error
succ_x = []
succ_err = []
for i in range(len(lmax_values) - 1):
    l1 = lmax_values[i]
    l2 = lmax_values[i + 1]
    succ_x.append(l1)
    succ_err.append(relative_error_percent(S_curves[l1], S_curves[l2]))

succ_x = np.array(succ_x)
succ_err = np.array(succ_err)

# Reference error vs lmax_ref
ref_x = []
ref_err = []
for lmax in lmax_values[:-1]:
    ref_x.append(lmax)
    ref_err.append(relative_error_percent(S_curves[lmax], S_curves[lmax_ref]))

ref_x = np.array(ref_x)
ref_err = np.array(ref_err)

print("\nSuccessive truncation errors [%]:")
for x, e in zip(succ_x, succ_err):
    print(f"  lmax={x:3d} -> next : {e:.6f}%")

print(f"\nReference truncation errors [%] relative to lmax_ref = {lmax_ref}")
for x, e in zip(ref_x, ref_err):
    print(f"  lmax={x:3d} -> ref  : {e:.6f}%")

# =========================================================
# Figure 2
# =========================================================
fig, axes = plt.subplots(2, 1, figsize=(7, 9))

# Top panel
for lmax in lmax_values:
    axes[0].plot(
        n_values,
        S_curves[lmax],
        marker="o",
        label=fr"$\ell_{{\max}}={lmax}$"
    )

axes[0].set_xlabel(r"$n$")
axes[0].set_ylabel(r"$S_{\rm tot}(n)$")
axes[0].legend()

# Bottom panel
axes[1].loglog(
    succ_x, succ_err,
    marker="o",
    label=r"successive error: $\ell_{\max}\to$ next"
)
axes[1].loglog(
    ref_x, ref_err,
    marker="s",
    linestyle="--",
    label=fr"reference error vs $\ell_{{\max}}={lmax_ref}$"
)

axes[1].set_xlabel(r"$\ell_{\max}$")
axes[1].set_ylabel("maximum relative error [%]")
axes[1].legend()

fig.tight_layout()
fig.savefig(outdir / "figure2_truncation_test_final.pdf")
fig.savefig(outdir / "figure2_truncation_test_final.png", dpi=300)
plt.show()

# =========================================================
# Angular-tail diagnostic
# =========================================================
ells = np.arange(lmax_ref + 1)
S_ell_tail = np.array(
    [subsystem_entropy_from_covariances(X_cache[ell], P_cache[ell], n_tail) for ell in ells],
    dtype=float
)

weighted_tail = (2 * ells + 1) * S_ell_tail
cumulative_tail = np.cumsum(weighted_tail)
cumulative_fraction = cumulative_tail / cumulative_tail[-1]

fig, axes = plt.subplots(2, 1, figsize=(7, 8), sharex=True)

axes[0].semilogy(ells, np.clip(S_ell_tail, 1e-16, None), marker="o", ms=3)
axes[0].set_ylabel(r"$S_\ell$")
axes[0].set_title(fr"Angular-sector entropy at fixed subsystem size $n={n_tail}$")

axes[1].semilogy(
    ells,
    np.clip(weighted_tail, 1e-16, None),
    marker="o",
    ms=3,
    label=r"$(2\ell+1)S_\ell$"
)
axes[1].set_xlabel(r"$\ell$")
axes[1].set_ylabel("weighted contribution")
axes[1].legend()

fig.tight_layout()
fig.savefig(outdir / "angular_tail_diagnostic_final.pdf")
fig.savefig(outdir / "angular_tail_diagnostic_final.png", dpi=300)
plt.show()

# =========================================================
# Cumulative angular fraction
# =========================================================
fig, ax = plt.subplots(figsize=(7, 5))
ax.plot(ells, cumulative_fraction, marker="o", ms=3)
ax.set_xlabel(r"$\ell$")
ax.set_ylabel("cumulative fraction of total")
ax.set_title(fr"Cumulative angular contribution at fixed $n={n_tail}$")
fig.tight_layout()
fig.savefig(outdir / "angular_tail_cumulative_fraction_final.pdf")
fig.savefig(outdir / "angular_tail_cumulative_fraction_final.png", dpi=300)
plt.show()

# =========================================================
# Optional: print final convergence statement
# =========================================================
final_ref_error = ref_err[-1]
print(f"\nFinal reference error (largest non-reference vs reference) = {final_ref_error:.6f}%")

# =========================================================
# Save data
# =========================================================
save_dict = {
    "a": np.array([a]),
    "N": np.array([N]),
    "n_values": n_values,
    "n_tail": np.array([n_tail]),
    "lmax_values": np.array(lmax_values),
    "succ_x": succ_x,
    "succ_err": succ_err,
    "ref_x": ref_x,
    "ref_err": ref_err,
    "ells": ells,
    "S_ell_tail": S_ell_tail,
    "weighted_tail": weighted_tail,
    "cumulative_fraction": cumulative_fraction,
}

for lmax in lmax_values:
    save_dict[f"S_lmax_{lmax}"] = S_curves[lmax]

np.savez(outdir / "flat_convergence_and_tail_data_final.npz", **save_dict)

print("\nSaved:")
print(" - figure2_truncation_test_final.pdf/png")
print(" - angular_tail_diagnostic_final.pdf/png")
print(" - angular_tail_cumulative_fraction_final.pdf/png")
print(" - flat_convergence_and_tail_data_final.npz")
print(f"\nTotal runtime: {time.time() - t0:.2f} s")
```

```python
import time
import numpy as np
import matplotlib.pyplot as plt
from scipy.linalg import eigh
from pathlib import Path

# =========================================================
# Plot style
# =========================================================
plt.rcParams.update({
    "figure.figsize": (12, 5),
    "axes.labelsize": 15,
    "axes.titlesize": 15,
    "legend.fontsize": 11,
    "xtick.labelsize": 12,
    "ytick.labelsize": 12,
    "lines.linewidth": 2.2,
    "lines.markersize": 7,
    "savefig.bbox": "tight",
    "axes.grid": False,
})

outdir = Path("paper_figures")
outdir.mkdir(exist_ok=True)
print("Saving figures in:", outdir.resolve())

t0 = time.time()

# =========================================================
# Numerical parameters
# =========================================================
a = 1.0
N = 120
r_flat = a * np.arange(1, N + 1)

# Safe subsystem range
n_values = np.arange(5, 31, 2)   # 5, 7, ..., 29

# Best available angular truncation
lmax_use = 280

print("a =", a)
print("N =", N)
print("n_values =", n_values)
print("lmax_use =", lmax_use)

# =========================================================
# Flat-space metric and coefficient builders
# =========================================================
def flat_metric_f(r):
    return np.ones_like(r)

def flat_metric_g(r):
    return np.ones_like(r)

def build_static_FGU(r, ell, f, g):
    ff = f(r)
    gg = g(r)

    if np.any(ff <= 0) or np.any(gg <= 0):
        raise ValueError("Metric functions must stay positive on the chosen lattice.")

    F = r**2 * np.sqrt(ff * gg)
    G = (ff * gg)**0.25 / r
    U = ell * (ell + 1.0) * ff / r**2
    return F, G, U

def build_K_from_coeffs(F, G, U, a):
    Nloc = len(U)
    K = np.zeros((Nloc, Nloc), dtype=float)

    # Off-diagonal
    for j in range(Nloc - 1):
        off = -F[j] * G[j] * G[j + 1] / a**2
        K[j, j + 1] = off
        K[j + 1, j] = off

    # Diagonal
    if Nloc >= 2:
        K[0, 0] = U[0] + F[0] * G[0]**2 / a**2
        K[-1, -1] = U[-1] + F[-2] * G[-1]**2 / a**2
        for j in range(1, Nloc - 1):
            K[j, j] = U[j] + (F[j] + F[j - 1]) * G[j]**2 / a**2
    else:
        K[0, 0] = U[0]

    K = 0.5 * (K + K.T)

    # Numerical stabilization if needed
    evals = np.linalg.eigvalsh(K)
    if evals.min() <= 0:
        K += np.eye(Nloc) * (1e-12 - evals.min() + 1e-12)

    return K

# =========================================================
# Ground-state covariances and subsystem entropy
# =========================================================
def ground_state_covariances(K):
    w, v = eigh(K)
    w = np.clip(w, 1e-15, None)
    sqrtK = (v * np.sqrt(w)) @ v.T
    invsqrtK = (v * (1.0 / np.sqrt(w))) @ v.T
    X = 0.5 * invsqrtK
    P = 0.5 * sqrtK
    return X, P

def subsystem_entropy_from_covariances(X, P, n_sub):
    XA = X[:n_sub, :n_sub]
    PA = P[:n_sub, :n_sub]

    eigvals = np.linalg.eigvals(XA @ PA)
    eigvals = np.real_if_close(eigvals, tol=1000)
    eigvals = np.real(eigvals)
    eigvals = np.clip(eigvals, 0.25, None)

    nus = np.sqrt(eigvals)
    nus = np.clip(nus, 0.5 + 1e-14, None)

    S = np.sum(
        (nus + 0.5) * np.log(nus + 0.5)
        - (nus - 0.5) * np.log(nus - 0.5)
    )
    return float(np.real(S))

# =========================================================
# Build and cache K^(ell), X^(ell), P^(ell)
# =========================================================
def flat_K(ell):
    F, G, U = build_static_FGU(r_flat, ell, flat_metric_f, flat_metric_g)
    return build_K_from_coeffs(F, G, U, a)

K_cache = {}
X_cache = {}
P_cache = {}

t_build = time.time()
for ell in range(lmax_use + 1):
    K = flat_K(ell)
    X, P = ground_state_covariances(K)
    K_cache[ell] = K
    X_cache[ell] = X
    P_cache[ell] = P
print(f"Cached K, X, P up to ell = {lmax_use} in {time.time() - t_build:.2f} s")

# =========================================================
# Build final entropy curve
# =========================================================
S_ell_cache = {}
for ell in range(lmax_use + 1):
    X = X_cache[ell]
    P = P_cache[ell]
    S_ell_cache[ell] = np.array(
        [subsystem_entropy_from_covariances(X, P, int(n)) for n in n_values],
        dtype=float
    )

S_tot = np.zeros_like(n_values, dtype=float)
for ell in range(lmax_use + 1):
    S_tot += (2 * ell + 1) * S_ell_cache[ell]

print("Built final flat-space entropy curve.")

# =========================================================
# Derived quantities
# =========================================================
R_flat = n_values * a
A_flat = 4.0 * np.pi * R_flat**2
ratio_flat = S_tot / A_flat

# Choose the large-R region for fitting: use the second half of the data
fit_start = len(A_flat) // 2
A_fit_region = A_flat[fit_start:]
S_fit_region = S_tot[fit_start:]

# Linear fit in the large-R region: S_tot = alpha * A + beta
alpha_fit, beta_fit = np.polyfit(A_fit_region, S_fit_region, 1)
S_fit = alpha_fit * A_flat + beta_fit

# Mean and scatter of the ratio over the large-R region
ratio_largeR = ratio_flat[fit_start:]
ratio_mean = np.mean(ratio_largeR)
ratio_std = np.std(ratio_largeR)

# Residuals of the large-R fit
fit_residual_percent = 100.0 * np.max(np.abs((S_fit_region - (alpha_fit * A_fit_region + beta_fit)) / S_fit_region))

print("\nLarge-R fit results:")
print(f"  alpha = {alpha_fit:.8e}")
print(f"  beta  = {beta_fit:.8e}")
print(f"  mean[S/(4πR²)] over large-R region = {ratio_mean:.8e}")
print(f"  std [S/(4πR²)] over large-R region = {ratio_std:.8e}")
print(f"  max fit residual in large-R region = {fit_residual_percent:.6f}%")

# =========================================================
# Figure 3: paper-ready flat-space validation
# =========================================================
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# ---------------------------------------------------------
# Left panel: S_tot vs area
# ---------------------------------------------------------
axes[0].plot(
    A_flat, S_tot,
    marker="o",
    label="numerical result"
)
axes[0].plot(
    A_flat, S_fit,
    linestyle="--",
    label="large-$R$ linear fit"
)

# Highlight fit region
axes[0].axvspan(
    A_fit_region.min(),
    A_fit_region.max(),
    alpha=0.08,
    color="gray"
)

axes[0].set_xlabel(r"$A$")
axes[0].set_ylabel(r"$S_{\mathrm{tot}}$")
axes[0].legend()

# ---------------------------------------------------------
# Right panel: area-law diagnostic
# ---------------------------------------------------------
axes[1].plot(
    R_flat, ratio_flat,
    marker="o",
    label=r"$S_{\mathrm{tot}}/(4\pi R^2)$"
)

# Mean large-R guide line
axes[1].axhline(
    ratio_mean,
    linestyle="--",
    linewidth=1.8,
    label="mean over large-$R$ region"
)

# Highlight fit / large-R region
axes[1].axvspan(
    R_flat[fit_start],
    R_flat[-1],
    alpha=0.08,
    color="gray"
)

axes[1].set_xlabel(r"$R$")
axes[1].set_ylabel(r"$S$")
axes[1].legend()

fig.tight_layout()
fig.savefig(outdir / "figure3_flat_validation_paper_ready.pdf")
fig.savefig(outdir / "figure3_flat_validation_paper_ready.png", dpi=300)
plt.show()

# =========================================================
# Save numerical data
# =========================================================
save_dict = {
    "a": np.array([a]),
    "N": np.array([N]),
    "lmax_use": np.array([lmax_use]),
    "n_values": n_values,
    "R_flat": R_flat,
    "A_flat": A_flat,
    "S_tot": S_tot,
    "ratio_flat": ratio_flat,
    "fit_start_index": np.array([fit_start]),
    "alpha_fit": np.array([alpha_fit]),
    "beta_fit": np.array([beta_fit]),
    "ratio_mean_largeR": np.array([ratio_mean]),
    "ratio_std_largeR": np.array([ratio_std]),
    "fit_residual_percent": np.array([fit_residual_percent]),
}

np.savez(outdir / "figure3_flat_validation_paper_ready_data.npz", **save_dict)

print("\nSaved:")
print(" - figure3_flat_validation_paper_ready.pdf/png")
print(" - figure3_flat_validation_paper_ready_data.npz")
print(f"\nTotal runtime: {time.time() - t0:.2f} s")
```

```python
import time
import numpy as np
import matplotlib.pyplot as plt
from scipy.linalg import eigh
from pathlib import Path

# =========================================================
# Plot style
# =========================================================
plt.rcParams.update({
    "figure.figsize": (12, 5),
    "axes.labelsize": 15,
    "axes.titlesize": 15,
    "legend.fontsize": 11,
    "xtick.labelsize": 12,
    "ytick.labelsize": 12,
    "lines.linewidth": 2.2,
    "lines.markersize": 7,
    "savefig.bbox": "tight",
    "axes.grid": False,
})

outdir = Path("paper_figures")
outdir.mkdir(exist_ok=True)
print("Saving figures in:", outdir.resolve())

t0 = time.time()

# =========================================================
# Numerical parameters
# =========================================================
a = 1.0
N = 120
M = 10.0

# Safe subsystem range
n_values = np.arange(5, 31, 2)

# Use a strong but still manageable truncation
# Increase to 280 later if runtime is acceptable
lmax_use = 200

print("a =", a)
print("N =", N)
print("M =", M)
print("n_values =", n_values)
print("lmax_use =", lmax_use)

# =========================================================
# Metric functions
# =========================================================
def schwarzschild_f(r, M):
    return 1.0 - 2.0 * M / r

def schwarzschild_g(r, M):
    return 1.0 - 2.0 * M / r

def schwarzschild_iso_C(rho, M):
    return (1.0 - M / (2.0 * rho)) / (1.0 + M / (2.0 * rho))

def schwarzschild_iso_B(rho, M):
    return (1.0 + M / (2.0 * rho))**2

def schwarzschild_iso_areal_radius(rho, M):
    return rho * (1.0 + M / (2.0 * rho))**2

# =========================================================
# Coefficient builders
# =========================================================
def build_static_FGU(r, ell, f, g):
    ff = f(r)
    gg = g(r)

    if np.any(ff <= 0) or np.any(gg <= 0):
        raise ValueError("Metric functions must stay positive on the chosen lattice.")

    F = r**2 * np.sqrt(ff * gg)
    G = (ff * gg)**0.25 / r
    U = ell * (ell + 1.0) * ff / r**2
    return F, G, U

def build_isotropic_lattice_coeffs(rho, ell, Cfunc, Bfunc, M, m_scalar=0.0, xi=0.0):
    """
    Vacuum isotropic Schwarzschild outside the source has R = 0.
    """
    C = Cfunc(rho, M)
    B = Bfunc(rho, M)
    R = np.zeros_like(rho)

    if np.any(C <= 0) or np.any(B <= 0):
        raise ValueError("C and B must stay positive on the chosen isotropic lattice.")

    CB = C * B
    dCB = np.gradient(CB, rho, edge_order=2)

    F_iso = C * B
    W_iso = np.sqrt(C / B**3)
    V_iso = (
        (C**2 / B**3) * (dCB / rho)
        + (C**2 / B**2) * ell * (ell + 1.0) / rho**2
        + C**2 * (m_scalar**2 + xi * R)
    )
    return F_iso, W_iso, V_iso

# =========================================================
# Lattice matrix builder
# =========================================================
def build_K_from_coeffs(F, G, U, a):
    Nloc = len(U)
    K = np.zeros((Nloc, Nloc), dtype=float)

    # Off-diagonal
    for j in range(Nloc - 1):
        off = -F[j] * G[j] * G[j + 1] / a**2
        K[j, j + 1] = off
        K[j + 1, j] = off

    # Diagonal
    if Nloc >= 2:
        K[0, 0] = U[0] + F[0] * G[0]**2 / a**2
        K[-1, -1] = U[-1] + F[-2] * G[-1]**2 / a**2
        for j in range(1, Nloc - 1):
            K[j, j] = U[j] + (F[j] + F[j - 1]) * G[j]**2 / a**2
    else:
        K[0, 0] = U[0]

    K = 0.5 * (K + K.T)

    evals = np.linalg.eigvalsh(K)
    if evals.min() <= 0:
        K += np.eye(Nloc) * (1e-12 - evals.min() + 1e-12)

    return K

# =========================================================
# Ground-state covariances and subsystem entropy
# =========================================================
def ground_state_covariances(K):
    w, v = eigh(K)
    w = np.clip(w, 1e-15, None)
    sqrtK = (v * np.sqrt(w)) @ v.T
    invsqrtK = (v * (1.0 / np.sqrt(w))) @ v.T
    X = 0.5 * invsqrtK
    P = 0.5 * sqrtK
    return X, P

def subsystem_entropy_from_covariances(X, P, n_sub):
    XA = X[:n_sub, :n_sub]
    PA = P[:n_sub, :n_sub]

    eigvals = np.linalg.eigvals(XA @ PA)
    eigvals = np.real_if_close(eigvals, tol=1000)
    eigvals = np.real(eigvals)
    eigvals = np.clip(eigvals, 0.25, None)

    nus = np.sqrt(eigvals)
    nus = np.clip(nus, 0.5 + 1e-14, None)

    S = np.sum(
        (nus + 0.5) * np.log(nus + 0.5)
        - (nus - 0.5) * np.log(nus - 0.5)
    )
    return float(np.real(S))

# =========================================================
# Radial lattices
# =========================================================
# Curvature coordinates: start safely outside horizon r = 2M
r_curv = (2.5 * M) + a * np.arange(N)

# Isotropic coordinates: start safely outside rho = M/2
rho_iso = (0.6 * M) + a * np.arange(N)
r_areal_iso = schwarzschild_iso_areal_radius(rho_iso, M)

print("Curvature-coordinate range:", r_curv[0], "to", r_curv[-1])
print("Isotropic-coordinate range:", rho_iso[0], "to", rho_iso[-1])
print("Areal-radius range from isotropic grid:", r_areal_iso[0], "to", r_areal_iso[-1])

# =========================================================
# Build and cache curvature-coordinate sector data
# =========================================================
X_cache_curv = {}
P_cache_curv = {}

t_build_curv = time.time()
for ell in range(lmax_use + 1):
    F, G, U = build_static_FGU(
        r_curv, ell,
        lambda r: schwarzschild_f(r, M),
        lambda r: schwarzschild_g(r, M)
    )
    K = build_K_from_coeffs(F, G, U, a)
    X, P = ground_state_covariances(K)
    X_cache_curv[ell] = X
    P_cache_curv[ell] = P
print(f"Built curvature-coordinate sector caches in {time.time() - t_build_curv:.2f} s")

# =========================================================
# Build and cache isotropic-coordinate sector data
# =========================================================
X_cache_iso = {}
P_cache_iso = {}

t_build_iso = time.time()
for ell in range(lmax_use + 1):
    F_iso, W_iso, V_iso = build_isotropic_lattice_coeffs(
        rho_iso, ell,
        schwarzschild_iso_C, schwarzschild_iso_B, M,
        m_scalar=0.0, xi=0.0
    )
    K_iso = build_K_from_coeffs(F_iso, W_iso, V_iso, a)
    X_iso, P_iso = ground_state_covariances(K_iso)
    X_cache_iso[ell] = X_iso
    P_cache_iso[ell] = P_iso
print(f"Built isotropic-coordinate sector caches in {time.time() - t_build_iso:.2f} s")

# =========================================================
# Build entropy curves
# =========================================================
S_tot_curv = np.zeros_like(n_values, dtype=float)
S_tot_iso = np.zeros_like(n_values, dtype=float)

for ell in range(lmax_use + 1):
    weight = 2 * ell + 1

    S_curv_ell = np.array(
        [subsystem_entropy_from_covariances(X_cache_curv[ell], P_cache_curv[ell], int(n)) for n in n_values],
        dtype=float
    )
    S_iso_ell = np.array(
        [subsystem_entropy_from_covariances(X_cache_iso[ell], P_cache_iso[ell], int(n)) for n in n_values],
        dtype=float
    )

    S_tot_curv += weight * S_curv_ell
    S_tot_iso += weight * S_iso_ell

print("Built entropy curves for both coordinate systems.")

# =========================================================
# Convert subsystem labels to areal radii
# =========================================================
R_curv = r_curv[n_values - 1]
R_iso_areal = r_areal_iso[n_values - 1]

# =========================================================
# Build common overlap region and relative difference
# =========================================================
R_min = max(R_curv.min(), R_iso_areal.min())
R_max = min(R_curv.max(), R_iso_areal.max())

mask_curv = (R_curv >= R_min) & (R_curv <= R_max)
R_common = R_curv[mask_curv]
S_curv_common = S_tot_curv[mask_curv]

# Interpolate isotropic entropy onto the curvature areal-radius points
S_iso_interp = np.interp(R_common, R_iso_areal, S_tot_iso)

# Symmetric relative difference
rel_diff_percent = 100.0 * np.abs(S_curv_common - S_iso_interp) / (
    0.5 * (S_curv_common + S_iso_interp)
)

mean_rel_diff = np.mean(rel_diff_percent)
max_rel_diff = np.max(rel_diff_percent)

print("\nCommon areal-radius overlap:")
print(f"  R_min = {R_min:.6f}")
print(f"  R_max = {R_max:.6f}")
print(f"  mean relative difference = {mean_rel_diff:.6f}%")
print(f"  max  relative difference = {max_rel_diff:.6f}%")

# =========================================================
# Figure 4: Schwarzschild entropy comparison
# =========================================================
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Left panel: entropy vs areal radius
axes[0].plot(
    R_curv, S_tot_curv,
    marker="o",
    label="curvature coordinates"
)
axes[0].plot(
    R_iso_areal, S_tot_iso,
    marker="s",
    label="isotropic coordinates"
)

# Highlight overlap region
axes[0].axvspan(R_min, R_max, alpha=0.08, color="gray")

axes[0].set_xlabel(r"areal radius $R$")
axes[0].set_ylabel(r"$S_{\mathrm{tot}}$")
axes[0].set_title("Schwarzschild entropy comparison")
axes[0].legend()

# Right panel: relative difference over common interval
axes[1].plot(
    R_common, rel_diff_percent,
    marker="o",
    label="relative difference"
)
axes[1].axhline(
    mean_rel_diff,
    linestyle="--",
    linewidth=1.8,
    label="mean difference"
)
axes[1].set_xlabel(r"areal radius $R$")
axes[1].set_ylabel("relative difference [%]")
axes[1].set_title("Coordinate-comparison residual")
axes[1].legend()

fig.tight_layout()
fig.savefig(outdir / "figure4_schwarzschild_coordinate_comparison.pdf")
fig.savefig(outdir / "figure4_schwarzschild_coordinate_comparison.png", dpi=300)
plt.show()

# =========================================================
# Save data
# =========================================================
save_dict = {
    "a": np.array([a]),
    "N": np.array([N]),
    "M": np.array([M]),
    "lmax_use": np.array([lmax_use]),
    "n_values": n_values,
    "R_curv": R_curv,
    "R_iso_areal": R_iso_areal,
    "S_tot_curv": S_tot_curv,
    "S_tot_iso": S_tot_iso,
    "R_common": R_common,
    "S_curv_common": S_curv_common,
    "S_iso_interp": S_iso_interp,
    "rel_diff_percent": rel_diff_percent,
    "mean_rel_diff": np.array([mean_rel_diff]),
    "max_rel_diff": np.array([max_rel_diff]),
}

np.savez(outdir / "figure4_schwarzschild_coordinate_comparison_data.npz", **save_dict)

print("\nSaved:")
print(" - figure4_schwarzschild_coordinate_comparison.pdf/png")
print(" - figure4_schwarzschild_coordinate_comparison_data.npz")
print(f"\nTotal runtime: {time.time() - t0:.2f} s")
```
