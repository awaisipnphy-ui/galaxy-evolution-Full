```python
# Require positive g and r flux
flux_mask = (lrg["FLUX_G"] > 0) & (lrg["FLUX_R"] > 0)

lrg = lrg[flux_mask]

print("After photometry cut:", len(lrg))
lrg["G_MINUS_R"] = -2.5 * np.log10(lrg["FLUX_G"] / lrg["FLUX_R"])
print(lrg["G_MINUS_R"].min(), lrg["G_MINUS_R"].max())
print(lrg["G_MINUS_R"][:10])
```

```python
# Select first redshift slice
zmin, zmax = 0.4, 0.6
slice1 = lrg[(lrg["Z_QN"] > zmin) & (lrg["Z_QN"] <= zmax)]

colors = slice1["G_MINUS_R"]
print("Galaxies in slice:", len(colors))
```

```python
import numpy as np

# Histogram
nbins = 100
hist, bin_edges = np.histogram(colors, bins=nbins, density=True)

# Bin centers
bin_centers = 0.5 * (bin_edges[:-1] + bin_edges[1:])
# Cumulative sums
weight1 = np.cumsum(hist)
weight2 = np.cumsum(hist[::-1])[::-1]

# Avoid division by zero
valid = (weight1 > 0) & (weight2 > 0)

# Cumulative means
mean1 = np.cumsum(hist * bin_centers) / weight1
mean2 = (np.cumsum((hist * bin_centers)[::-1]) / weight2[::-1])[::-1]

# Inter-class variance
sigma_b2 = weight1[:-1] * weight2[1:] * (mean1[:-1] - mean2[1:])**2

# Otsu threshold index
idx = np.argmax(sigma_b2)

threshold = bin_centers[idx]

print("Otsu threshold (g-r) for 0.4<z≤0.6:", threshold)
import matplotlib.pyplot as plt

plt.figure(figsize=(7,5))
plt.hist(colors, bins=100, density=True, alpha=0.6)
plt.axvline(threshold, color='red', linestyle='--', label='Otsu threshold')
plt.xlabel("(g − r)")
plt.ylabel("Probability density")
plt.title("LRG color distribution (0.4 < z ≤ 0.6)")
plt.legend()
plt.show()
```

```python
# Second redshift slice
zmin, zmax = 0.6, 0.8
slice2 = lrg[(lrg["Z_QN"] > zmin) & (lrg["Z_QN"] <= zmax)]

colors2 = slice2["G_MINUS_R"]

print("Galaxies in slice 0.6 < z ≤ 0.8:", len(colors2))
import numpy as np

nbins = 100
hist2, bin_edges2 = np.histogram(colors2, bins=nbins, density=True)

bin_centers2 = 0.5 * (bin_edges2[:-1] + bin_edges2[1:])
# Cumulative weights
w1 = np.cumsum(hist2)
w2 = np.cumsum(hist2[::-1])[::-1]

valid = (w1 > 0) & (w2 > 0)

# Cumulative means
m1 = np.cumsum(hist2 * bin_centers2) / w1
m2 = (np.cumsum((hist2 * bin_centers2)[::-1]) / w2[::-1])[::-1]

# Inter-class variance
sigma_b2 = w1[:-1] * w2[1:] * (m1[:-1] - m2[1:])**2

# Otsu threshold
idx2 = np.argmax(sigma_b2)
threshold2 = bin_centers2[idx2]

print("Otsu threshold (g-r) for 0.6<z≤0.8:", threshold2)
import matplotlib.pyplot as plt

plt.figure(figsize=(7,5))
plt.hist(colors2, bins=100, density=True, alpha=0.6)
plt.axvline(threshold2, color='red', linestyle='--', label='Otsu threshold')
plt.xlabel("(g − r)")
plt.ylabel("Probability density")
plt.title("LRG color distribution (0.6 < z ≤ 0.8)")
plt.legend()
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import os
from scipy.interpolate import interp1d

# ============================================================
# STEP 1: LOAD AND CLEAN DATA
# ============================================================
print("=== Loading and cleaning data ===")

project_path = "/home/qarni/galaxy_evolution_project/LRG"
os.chdir(project_path)

df = pd.read_csv("LRG_Stage2_Color_StellarMass_FINAL.csv")
print(f"Loaded {len(df):,} rows")

df_clean = df.dropna(subset=["Z_QN", "LOG_MASS_CG", "G_MINUS_R"]).copy()
print(f"After cleaning: {len(df_clean):,} rows")

# Redshift bins (as in your analysis)
z_bins = [(0.6, 0.7), (0.7, 0.8), (0.8, 0.9)]

df_filtered = df_clean[
    (df_clean["Z_QN"] >= 0.4) &
    (df_clean["Z_QN"] < 1.0)
].copy()

print(f"In redshift range 0.4–1.0: {len(df_filtered):,} galaxies")

# ============================================================
# STEP 2: OTSU THRESHOLD (EQS. 3.1–3.9, PAPER-FAITHFUL)
# ============================================================
def paper_otsu_threshold(colors, M=256, color_range=(0.5, 3.0)):
    """
    Otsu thresholding following Pandey et al. (Eqs. 3.1–3.9).
    """
    if len(colors) < 20:
        return None

    hist, bin_edges = np.histogram(colors, bins=M, range=color_range)
    N = hist.sum()
    if N == 0:
        return None

    pi = hist / N                                  # p_i
    xi = 0.5 * (bin_edges[:-1] + bin_edges[1:])   # x_i

    w = np.cumsum(pi)                             # w(k)
    mu_k = np.cumsum(xi * pi)                     # μ_k
    mu_T = mu_k[-1]                               # μ_T

    sigma_bc_max = -np.inf
    threshold = None

    for k in range(M - 1):
        if w[k] == 0 or w[k] == 1:
            continue

        P_BC = w[k]
        P_RS = 1.0 - w[k]

        mu_BC = mu_k[k] / P_BC
        mu_RS = (mu_T - mu_k[k]) / P_RS

        sigma_bc = P_BC * P_RS * (mu_BC - mu_RS)**2

        if sigma_bc > sigma_bc_max:
            sigma_bc_max = sigma_bc
            threshold = xi[k]

    return threshold

# ============================================================
# STEP 3: MASS-DEPENDENT OTSU
# ============================================================
def mass_dependent_otsu(df, mass_bins, min_count=200):
    mass_centers, thresholds = [], []

    for mmin, mmax in zip(mass_bins[:-1], mass_bins[1:]):
        sub = df[
            (df["LOG_MASS_CG"] >= mmin) &
            (df["LOG_MASS_CG"] <  mmax)
        ]

        if len(sub) < min_count:
            continue

        t = paper_otsu_threshold(sub["G_MINUS_R"].values)
        if t is not None:
            mass_centers.append(0.5 * (mmin + mmax))
            thresholds.append(t)

    return np.array(mass_centers), np.array(thresholds)

# ============================================================
# STEP 4: PLOTTING (PANDEY STYLE)
# ============================================================
mass_bins = np.arange(10.5, 12.5, 0.2)

fig, axes = plt.subplots(1, 3, figsize=(16, 6), sharex=True, sharey=True)

summary = []

for i, (zmin, zmax) in enumerate(z_bins):
    ax = axes[i]

    dfz = df_filtered[
        (df_filtered["Z_QN"] >= zmin) &
        (df_filtered["Z_QN"] <  zmax)
    ].copy()

    print(f"z = {zmin}–{zmax}: {len(dfz):,} galaxies")

    mcen, otsu_vals = mass_dependent_otsu(dfz, mass_bins, min_count=150)
    if len(mcen) < 2:
        continue

    idx = np.argsort(mcen)
    mcen, otsu_vals = mcen[idx], otsu_vals[idx]

    f_div = interp1d(
        mcen, otsu_vals,
        kind="linear",
        bounds_error=False,
        fill_value=(otsu_vals[0], otsu_vals[-1])
    )

    divider = f_div(dfz["LOG_MASS_CG"].values)
    is_red = dfz["G_MINUS_R"].values >= divider

    blue = dfz[~is_red]
    red  = dfz[is_red]

    summary.append({
        "z": f"{zmin}-{zmax}",
        "total": len(dfz),
        "red": len(red),
        "blue": len(blue),
        "bins": len(mcen)
    })

    # ---------- SCATTER (LOW NOISE, HIGH TRANSPARENCY) ----------
    ax.scatter(
        blue["LOG_MASS_CG"], blue["G_MINUS_R"],
        s=0.6, alpha=0.08,
        color="royalblue", rasterized=True,
        label=f'Blue Cloud ({len(blue):,})'
    )

    ax.scatter(
        red["LOG_MASS_CG"], red["G_MINUS_R"],
        s=0.6, alpha=0.08,
        color="darkred", rasterized=True,
        label=f'Red Sequence ({len(red):,})'
    )

    # ---------- DIVIDER ----------
    mgrid = np.linspace(mcen.min(), mcen.max(), 300)
    ax.plot(mgrid, f_div(mgrid), color="black", lw=3, zorder=10)

    ax.scatter(
        mcen, otsu_vals,
        s=30, color="black",
        edgecolor="black", linewidth=1,
        zorder=11
    )

    ax.set_title(f"{zmin} ≤ z < {zmax}", fontsize=14)
    ax.set_xlim(10.5, 12.5)
    ax.set_ylim(0.5, 3.0)
    ax.grid(alpha=0.3, linestyle="--")

    if i == 0:
        ax.set_ylabel(r"$(g-r)$", fontsize=12)

    # FIXED: corrected set_xlabel line
    ax.set_xlabel(r"$\log_{10}(M_*/M_\odot)$", fontsize=12)

# ============================================================
# FINAL FORMATTING
# ============================================================

plt.tight_layout()
plt.savefig("LRG_mass_dependent_otsu_PANDEY_STYLE.png",
            dpi=400, bbox_inches="tight", facecolor="white")
plt.show()

# ============================================================
# SUMMARY TABLE
# ============================================================
print("\nSUMMARY")
print("z-bin   Total     Red     Blue    Red%   Blue%  MassBins")
for s in summary:
    # FIXED: proper indentation (4 spaces) and line continuation
    print(f"{s['z']:>6} {s['total']:>8,} {s['red']:>8,} {s['blue']:>8,} "
          f"{100*s['red']/s['total']:6.1f}% {100*s['blue']/s['total']:6.1f}% {s['bins']:>5}")
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib as mpl

# =========================
# Clean publication defaults
# =========================
mpl.rcParams.update({
    "figure.dpi": 150,
    "savefig.dpi": 500,
    "font.size": 12,
    "axes.labelsize": 13,
    "axes.titlesize": 14,
    "legend.fontsize": 11,
    "axes.linewidth": 1.1,
    "xtick.direction": "in",
    "ytick.direction": "in",
    "xtick.major.size": 6,
    "ytick.major.size": 6,
    "xtick.minor.size": 3,
    "ytick.minor.size": 3,
    "xtick.minor.visible": True,
    "ytick.minor.visible": True,
})

# ============================================================
# Assume df_unique is already loaded with columns:
# Z, env_name, LOG_MASS_CG, is_red (boolean)
# ============================================================

# ---- 4 redshift bins ----
z_min, z_max = df_unique["Z"].min(), df_unique["Z"].max()
z_bins = np.linspace(z_min, z_max, 5)  # 5 edges -> 4 bins
z_centers = [(z_bins[i] + z_bins[i + 1]) / 2 for i in range(len(z_bins) - 1)]

def compute_fractions(data: pd.DataFrame, z_bins, z_centers) -> pd.DataFrame:
    env_list = sorted(data["env_name"].unique())  # stable order
    rows = []

    for i, zc in enumerate(z_centers):
        z_low, z_high = z_bins[i], z_bins[i + 1]
        df_z = data[(data["Z"] >= z_low) & (data["Z"] < z_high)]

        for env in env_list:
            df_env = df_z[df_z["env_name"] == env]

            n_red = int(df_env["is_red"].sum())
            n_tot = int(len(df_env))
            n_blue = n_tot - n_red

            if n_tot > 0:
                rf = n_red / n_tot
                bf = n_blue / n_tot
                err_rf = np.sqrt(rf * (1 - rf) / n_tot)
                err_bf = np.sqrt(bf * (1 - bf) / n_tot)
            else:
                rf = bf = err_rf = err_bf = np.nan

            rows.append({
                "z_center": zc,
                "env": env,
                "rf": rf, "bf": bf,
                "err_rf": err_rf, "err_bf": err_bf,
                "n_tot": n_tot
            })

    return pd.DataFrame(rows)

# ---- Full sample ----
df_full = compute_fractions(df_unique, z_bins, z_centers)
df_full["sample"] = "full"

# ---- High-mass sample ----
high_mask = (df_unique["LOG_MASS_CG"] > 10.5) & (df_unique["LOG_MASS_CG"] < 12)
df_high = df_unique.loc[high_mask].copy()
df_high_res = compute_fractions(df_high, z_bins, z_centers)
df_high_res["sample"] = "high_mass"

# ---- Environment styles ----
environments = ["Void", "Sheet", "Filament", "Knot"]
env_styles = {
    "Void": {"color": "orange", "marker": "o", "linestyle": "--"},
    "Sheet": {"color": "green", "marker": "s", "linestyle": "-"},
    "Filament": {"color": "blue", "marker": "^", "linestyle": "-"},
    "Knot": {"color": "red", "marker": "D", "linestyle": "-"},
}

env_order = [e for e in ["Void", "Sheet", "Filament", "Knot"]
             if e in df_unique["env_name"].unique()]

datasets = [
    (df_full, "All masses"),
    (df_high_res, r"mass ($10.5 < \log_{10}(M_*/M_\odot) < 12$)")
]

# =========================
# Plot
# =========================
fig, axes = plt.subplots(2, 2, figsize=(12, 12), sharex=True, sharey="row")

for col, (data, title) in enumerate(datasets):
    # --- Top: RF ---
    ax_top = axes[0, col]
    ax_top.set_title(title)

    for env in env_order:
        d = data[data["env"] == env].sort_values("z_center")
        if d.empty:
            continue
        st = env_styles[env]
        ax_top.errorbar(
            d["z_center"], d["rf"], yerr=d["err_rf"],
            fmt=st["marker"], linestyle=st["linestyle"], color=st["color"],
            capsize=3, elinewidth=1.0, linewidth=1.6, markersize=6,
            label=env
        )

    ax_top.set_ylabel(r"RF $\left(\frac{n_R}{n_R+n_B}\right)$")
    ax_top.set_ylim(0, 1.05)

    # IMPORTANT: remove ONLY inside lines (grid)
    ax_top.grid(False)

    if col == 1:
        ax_top.legend(loc="best", frameon=True)

    # --- Bottom: BF ---
    ax_bot = axes[1, col]

    for env in env_order:
        d = data[data["env"] == env].sort_values("z_center")
        if d.empty:
            continue
        st = env_styles[env]
        ax_bot.errorbar(
            d["z_center"], d["bf"], yerr=d["err_bf"],
            fmt=st["marker"], linestyle=st["linestyle"], color=st["color"],
            capsize=3, elinewidth=1.0, linewidth=1.6, markersize=6,
            label=env
        )

    ax_bot.set_ylabel(r"BF $\left(\frac{n_B}{n_R+n_B}\right)$")
    ax_bot.set_xlabel("Z")
    ax_bot.set_ylim(0, 1.05)

    # IMPORTANT: remove ONLY inside lines (grid)
    ax_bot.grid(False)

# x-limits across bins
axes[1, 0].set_xlim(z_bins[0], z_bins[-1])

plt.tight_layout()
plt.savefig("red_blue_fractions_4bins_box_nogrid.png", dpi=500, bbox_inches="tight", pad_inches=0.02)
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Assume df_unique is already loaded with columns: Z, env_name, is_red, LOG_MASS_CG

# Redshift bins (same as before)
z_min, z_max = df_unique['Z'].min(), df_unique['Z'].max()
z_bins = np.linspace(z_min, z_max, 5)
z_centers = [(z_bins[i] + z_bins[i+1])/2 for i in range(4)]

# Function to compute RRF and RBF for a given sample (dataframe)
def compute_rrf_rbf(data, z_bins, z_centers):
    results = []
    for i, zc in enumerate(z_centers):
        z_low = z_bins[i]
        z_high = z_bins[i+1]
        mask_z = (data['Z'] >= z_low) & (data['Z'] < z_high)
        df_z = data[mask_z]
        
        # Count red and blue per environment
        env_counts = {}
        for env in data['env_name'].unique():
            mask_env = (df_z['env_name'] == env)
            df_env = df_z[mask_env]
            n_red = df_env['is_red'].sum()
            n_blue = (~df_env['is_red']).sum()
            env_counts[env] = {'n_red': n_red, 'n_blue': n_blue}
        
        # Totals over all environments
        total_red = sum(v['n_red'] for v in env_counts.values())
        total_blue = sum(v['n_blue'] for v in env_counts.values())
        
        for env in data['env_name'].unique():
            n_red = env_counts[env]['n_red']
            n_blue = env_counts[env]['n_blue']
            
            # RRF and RBF with errors
            if total_red > 0:
                rrf = n_red / total_red
                err_rrf = np.sqrt(rrf * (1 - rrf) / total_red)
            else:
                rrf, err_rrf = np.nan, np.nan
            
            if total_blue > 0:
                rbf = n_blue / total_blue
                err_rbf = np.sqrt(rbf * (1 - rbf) / total_blue)
            else:
                rbf, err_rbf = np.nan, np.nan
            
            results.append({
                'z_center': zc,
                'env': env,
                'rrf': rrf,
                'rbf': rbf,
                'err_rrf': err_rrf,
                'err_rbf': err_rbf,
                'n_red': n_red,
                'n_blue': n_blue,
                'total_red': total_red,
                'total_blue': total_blue
            })
    return pd.DataFrame(results)

# Full sample
rrf_full = compute_rrf_rbf(df_unique, z_bins, z_centers)
rrf_full['sample'] = 'full'

# High-mass sample
high_mask = (df_unique['LOG_MASS_CG'] > 10.5) & (df_unique['LOG_MASS_CG'] < 12)
df_high = df_unique[high_mask].copy()
rrf_high = compute_rrf_rbf(df_high, z_bins, z_centers)
rrf_high['sample'] = 'high_mass'

# Plotting
environments = ["Void", "Sheet", "Filament", "Knot"]
env_styles = {
    "Void": {"color": "orange", "marker": "o", "linestyle": "--"},
    "Sheet": {"color": "green", "marker": "s", "linestyle": "-"},
    "Filament": {"color": "blue", "marker": "^", "linestyle": "-"},
    "Knot": {"color": "red", "marker": "D", "linestyle": "-"},
}

fig, axes = plt.subplots(2, 2, figsize=(12, 12), sharex=True, sharey='row')
titles = ['All masses', r'mass ($10.5 < \log_{10}(M_*/M_\odot) < 12$)']
datasets = [rrf_full, rrf_high]

for col, (data, title) in enumerate(zip(datasets, titles)):
    # Top row: RRF
    ax_top = axes[0, col]
    ax_top.set_title(title)

    for env in env_styles.keys():
        d = data[data['env'] == env].sort_values('z_center')
        style = env_styles[env]
        if len(d) > 0:
            ax_top.errorbar(
                d['z_center'], d['rrf'], yerr=d['err_rrf'],
                fmt=style['marker'], color=style['color'],
                linestyle=style['linestyle'], label=env, capsize=3
            )

    ax_top.set_ylabel(r'RRF$_i$ $\left(\frac{(n_R)_i}{\sum_{j=1}^{4} (n_R)_j}\right)$')
    if col == 1:
        ax_top.legend(loc='best')

    # remove internal grid lines (keep full box)
    ax_top.grid(False)
    ax_top.set_ylim(0, 0.6)

    # Bottom row: RBF
    ax_bottom = axes[1, col]
    for env in env_styles.keys():
        d = data[data['env'] == env].sort_values('z_center')
        style = env_styles[env]
        if len(d) > 0:
            ax_bottom.errorbar(
                d['z_center'], d['rbf'], yerr=d['err_rbf'],
                fmt=style['marker'], color=style['color'],
                linestyle=style['linestyle'], label=env, capsize=3
            )

    ax_bottom.set_ylabel(r'RBF$_i$ $\left(\frac{(n_B)_i}{\sum_{j=1}^{4} (n_B)_j}\right)$')
    ax_bottom.set_xlabel('Z')

    # remove internal grid lines (keep full box)
    ax_bottom.grid(False)
    ax_bottom.set_ylim(0, 0.6)

plt.tight_layout()
plt.savefig('rrf_rbf_full_high.png', dpi=550, bbox_inches='tight', pad_inches=0.02)
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib as mpl

# =========================
# Publication-style defaults
# =========================
mpl.rcParams.update({
    "figure.dpi": 150,
    "savefig.dpi": 550,
    "font.size": 12,
    "axes.labelsize": 13,
    "axes.titlesize": 14,
    "legend.fontsize": 10,
    "axes.linewidth": 1.2,
    "xtick.direction": "in",
    "ytick.direction": "in",
    "xtick.major.size": 6,
    "ytick.major.size": 6,
    "xtick.minor.size": 3,
    "ytick.minor.size": 3,
    "xtick.minor.visible": True,
    "ytick.minor.visible": True,
})

# =========================
# Load data (if needed)
# =========================
try:
    df_unique
except NameError:
    df_unique = pd.read_csv("merged_galaxies.csv")
    if "is_red" not in df_unique.columns:
        threshold = 1.8
        df_unique["is_red"] = df_unique["G_MINUS_R"] > threshold
    # Ensure mass column name
    if "LOG_MASS_CG" not in df_unique.columns:
        if "logM" in df_unique.columns:
            df_unique.rename(columns={"logM": "LOG_MASS_CG"}, inplace=True)

df_unique = df_unique.copy()
df_unique["is_red"] = df_unique["is_red"].astype(bool)

# Redshift bins
z_bins = [(0.6, 0.7), (0.7, 0.8), (0.8, 0.9)]
z_centers = [(lo+hi)/2 for lo, hi in z_bins]
z_labels = [f"{lo:.1f}≤z<{hi:.1f}" for lo, hi in z_bins]

# Mass bins (from 10.5 to 12.0, step 0.1)
mass_bins = np.arange(10.5, 12.0, 0.1)
mass_centers = (mass_bins[:-1] + mass_bins[1:]) / 2

# Environments
environments = ["Void", "Sheet", "Filament", "Knot"]
env_styles = {
    "Void": {"color": "orange", "marker": "o", "linestyle": "--"},
    "Sheet": {"color": "green", "marker": "s", "linestyle": "-"},
    "Filament": {"color": "blue", "marker": "^", "linestyle": "-"},
    "Knot": {"color": "red", "marker": "D", "linestyle": "-"},
}

env_order = [e for e in environments if e in df_unique["env_name"].unique()]

# Storage for RBF only
results_rbf = []
min_count = 5  # minimum number of galaxies per mass bin for a point to be shown

for (z_low, z_high), zc in zip(z_bins, z_centers):
    df_z = df_unique[(df_unique["Z"] >= z_low) & (df_unique["Z"] < z_high)]
    
    # For each mass bin, compute totals over all environments
    for mi, (m_low, m_high) in enumerate(zip(mass_bins[:-1], mass_bins[1:])):
        df_mass = df_z[(df_z["LOG_MASS_CG"] >= m_low) & (df_z["LOG_MASS_CG"] < m_high)]
        total_blue = (~df_mass["is_red"]).sum()
        
        # Compute per environment
        for env in env_order:
            df_env = df_mass[df_mass["env_name"] == env]
            n_blue = (~df_env["is_red"]).sum()
            
            # RBF
            if total_blue >= min_count:
                rbf = n_blue / total_blue
                err_rbf = np.sqrt(rbf * (1 - rbf) / total_blue)
            else:
                rbf, err_rbf = np.nan, np.nan
            
            results_rbf.append({
                "z_center": zc,
                "z_label": f"{z_low:.1f}≤z<{z_high:.1f}",
                "env": env,
                "mass_center": mass_centers[mi],
                "rbf": rbf,
                "err_rbf": err_rbf,
                "n_blue": n_blue,
                "total_blue": total_blue
            })

df_rbf = pd.DataFrame(results_rbf)

# Plotting: one row, three columns (RBF only)
fig, axes = plt.subplots(1, 3, figsize=(12, 4), sharex=True, sharey=True)

for col, z_label in enumerate(z_labels):
    ax = axes[col]
    data = df_rbf[df_rbf["z_label"] == z_label]
    
    for env in env_order:
        d = data[(data["env"] == env) & (data["rbf"].notna())].sort_values("mass_center")
        if d.empty:
            continue
        st = env_styles[env]
        ax.errorbar(d["mass_center"], d["rbf"], yerr=d["err_rbf"],
                    fmt=st["marker"], color=st["color"], linestyle=st["linestyle"],
                    capsize=3, elinewidth=1.0, linewidth=1.6, markersize=5,
                    label=env)
    
    ax.set_title(z_label)
    ax.set_xlabel(r"$\log_{10}(M_*/M_\odot)$")
    ax.set_ylabel(r"RBF$_i$ $\left(\frac{(n_B)_i}{\sum_j (n_B)_j}\right)$")
    ax.set_xlim(10.5, 12.0)
    ax.set_ylim(0, 1.0)
    ax.grid(False)
    ax.tick_params(which="both", top=True, right=True)

# Legend on the first panel only
axes[0].legend(loc="upper left", frameon=True)

plt.tight_layout()
plt.savefig("rbf_vs_mass_three_zbins.png", dpi=550, bbox_inches="tight")
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib as mpl
import math

# =========================
# Publication-style defaults
# =========================
mpl.rcParams.update({
    "figure.dpi": 150,
    "savefig.dpi": 550,
    "font.size": 12,
    "axes.labelsize": 13,
    "axes.titlesize": 14,
    "legend.fontsize": 10,
    "axes.linewidth": 1.2,
    "xtick.direction": "in",
    "ytick.direction": "in",
    "xtick.major.size": 6,
    "ytick.major.size": 6,
    "xtick.minor.size": 3,
    "ytick.minor.size": 3,
    "xtick.minor.visible": True,
    "ytick.minor.visible": True,
})

# =========================
# Load data (if needed)
# =========================
try:
    df_unique
except NameError:
    df_unique = pd.read_csv("merged_galaxies.csv")

df_unique = df_unique.copy()

# ------------------------------------------------------------
# Choose colour column: prefer (u-r) if exists, else (g-r)
# ------------------------------------------------------------
if "U_MINUS_R" in df_unique.columns:
    colour_col = "U_MINUS_R"
    x_label = r"$(u-r)$ colour"
    x_min, x_max = 0.5, 3.2
elif "U_R" in df_unique.columns:
    colour_col = "U_R"
    x_label = r"$(u-r)$ colour"
    x_min, x_max = 0.5, 3.2
elif "G_MINUS_R" in df_unique.columns:
    colour_col = "G_MINUS_R"
    x_label = r"$(g-r)$ colour"
    x_min, x_max = 0.5, 3.0
else:
    raise KeyError("No colour column found. Expected U_MINUS_R / U_R / G_MINUS_R")

# Required columns
for c in ["Z", "env_name", colour_col]:
    if c not in df_unique.columns:
        raise KeyError(f"Missing column '{c}'. Available: {list(df_unique.columns)}")

# =========================
# Redshift bins (your ranges)
# =========================
z_bins = [(0.6, 0.7), (0.7, 0.8), (0.8, 0.9)]
z_titles = [f"{lo:.1f}≤z<{hi:.1f}" for lo, hi in z_bins]

# =========================
# Environments and styles
# =========================
env_order = ["Void", "Sheet", "Filament", "Knot"]
env_styles = {
    "Void":     dict(color="tab:orange", linestyle="-"),
    "Sheet":    dict(color="tab:green",  linestyle="--"),
    "Filament": dict(color="tab:blue",   linestyle="-."),
    "Knot":     dict(color="tab:red",    linestyle="-"),
}

existing_envs = [e for e in env_order if e in set(df_unique["env_name"].unique())]
if len(existing_envs) == 0:
    raise ValueError(f"No matching envs found. Available: {df_unique['env_name'].unique()}")

# =========================
# Histogram PDF settings
# =========================
bins = 40
edges = np.linspace(x_min, x_max, bins + 1)
min_count = 50  # skip noisy curves

# =========================
# Layout: based on number of z bins
# =========================
n_panels = len(z_bins)
ncols = 3 if n_panels >= 3 else n_panels
nrows = int(math.ceil(n_panels / ncols))

fig, axes = plt.subplots(
    nrows, ncols,
    figsize=(5.2 * ncols, 4.2 * nrows),
    sharex=True, sharey=True
)
axes = np.atleast_1d(axes).ravel()

# =========================
# Plot each redshift panel with all environments
# =========================
for i, ((zlo, zhi), title) in enumerate(zip(z_bins, z_titles)):
    ax = axes[i]
    df_z = df_unique[(df_unique["Z"] >= zlo) & (df_unique["Z"] < zhi)]

    for env in existing_envs:
        x = df_z.loc[df_z["env_name"] == env, colour_col].dropna().values
        if x.size < min_count:
            continue

        pdf, _ = np.histogram(x, bins=edges, density=True)
        ax.step(edges[:-1], pdf, where="post",
                linewidth=2.0, label=env, **env_styles[env])

    ax.set_title(title)
    ax.set_xlim(x_min, x_max)
    ax.grid(False)
    ax.tick_params(which="both", top=True, right=True)

    # Only left-most panels get y-label (NO overlap)
    if (i % ncols) == 0:
        ax.set_ylabel("PDF")

    # Only bottom row panels get x-label
    if i >= (n_panels - ncols):
        ax.set_xlabel(x_label)

# Hide unused axes
for j in range(n_panels, len(axes)):
    axes[j].axis("off")

# Legend (first panel like your current)
axes[0].legend(loc="upper right", frameon=True)

# Give extra space on the left so ticks + 'PDF' never collide
fig.subplots_adjust(left=0.08, bottom=0.12, wspace=0.10)

plt.savefig("colour_pdf_by_redshift_panels.png", dpi=550, bbox_inches="tight", pad_inches=0.02)
plt.show()
```
