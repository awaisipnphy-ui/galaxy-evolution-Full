```python
# =====================================================
#  STAGE 1 — DESI DR1 (LRG): Clean & Convert  ✅ ENDIAN-SAFE
# =====================================================
import pandas as pd, numpy as np, os
from astropy.io import fits
from astropy.cosmology import Planck18 as cosmo
import astropy.units as u

# -----------------------------------------------------
# Helper: read DESI .fits (for galaxies or randoms)
# -----------------------------------------------------
def read_basic(fits_path, nz_path=None, nmax=None):
    """
    Reads a DESI .fits catalog and extracts RA, DEC, and Z.
    Handles endian conversion and random z-assignment for random catalogs.
    """
    with fits.open(fits_path, memmap=True) as hdul:
        data = hdul[1].data
        cols = data.columns.names

        # Force all columns to native (little-endian)
        def to_native(arr):
            return np.array(arr, dtype=arr.dtype.newbyteorder("="))

        ra = to_native(data["RA"])
        dec = to_native(data["DEC"])

        # --- Detect redshift source ---
        if "Z_not4clus" in cols:
            z = to_native(data["Z_not4clus"])
            print(f"✅ Using Z_not4clus from {fits_path}")
        elif "Z" in cols:
            z = to_native(data["Z"])
            print(f"✅ Using Z from {fits_path}")
        elif nz_path is not None:
            print(f"⚙️  No Z column, sampling from n(z) in {nz_path}")
            nz = np.loadtxt(nz_path)
            z_bins, nz_counts = nz[:, 0], nz[:, 1]
            z_probs = nz_counts / nz_counts.sum()
            z = np.random.choice(z_bins, size=len(ra), p=z_probs)
        else:
            raise KeyError(f"No redshift info found for {fits_path}")

        # Assemble DataFrame
        df = pd.DataFrame({"RA": ra, "DEC": dec, "Z": z})

        # Optional subsample (for test runs)
        if nmax:
            df = df.sample(nmax, random_state=42)

    print(f"✅ Loaded {len(df):,} rows from {os.path.basename(fits_path)}")
    return df


# -----------------------------------------------------
# Load LRG galaxy and random catalogs
# -----------------------------------------------------
lrg_gal = read_basic("LRG_full_HPmapcut.dat.fits")
lrg_ran = read_basic("LRG_0_full_HPmapcut.ran.fits",
                     nz_path="LRG_full_HPmapcut_nz.txt")

# -----------------------------------------------------
# Compute α normalization for LRG sample
# -----------------------------------------------------
lrg_gal = lrg_gal.assign(SAMPLE="LRG")
lrg_ran = lrg_ran.assign(SAMPLE="LRG")

alpha = len(lrg_gal) / len(lrg_ran)
print(f"\nNormalization factor α = {alpha:.3e}")
print(f"{len(lrg_gal):,} LRG galaxies  |  {len(lrg_ran):,} LRG randoms")


# -----------------------------------------------------
# Convert RA, DEC, z → Cartesian coordinates (safe)
# -----------------------------------------------------
def add_cartesian(df):
    """
    Converts RA, DEC, Z → x, y, z [Mpc] in comoving coordinates.
    Filters invalid or nonphysical redshifts automatically.
    """
    # Ensure numeric type and endian-safe conversion
    df["Z"] = np.array(df["Z"], dtype=float)
    df = df[np.isfinite(df["Z"]) & (df["Z"] > 0)].copy()

    chi = cosmo.comoving_distance(df["Z"]).to(u.Mpc).value

    ra, dec = np.deg2rad(df["RA"]), np.deg2rad(df["DEC"])
    df["x"] = chi * np.cos(dec) * np.cos(ra)
    df["y"] = chi * np.cos(dec) * np.sin(ra)
    df["z"] = chi * np.sin(dec)

    print(f"✅ Converted {len(df):,} valid entries to (x, y, z)")
    return df


lrg_gal = add_cartesian(lrg_gal)
lrg_ran = add_cartesian(lrg_ran)


# -----------------------------------------------------
# Save cleaned catalogs
# -----------------------------------------------------
out_gal = "lrg_galaxies_xyz.csv"
out_ran = "lrg_randoms_xyz.csv"

lrg_gal.to_csv(out_gal, index=False)
lrg_ran.to_csv(out_ran, index=False)

print(f"\n✅ Saved cleaned catalogs:")
print(f"   → {out_gal} ({len(lrg_gal):,} galaxies)")
print(f"   → {out_ran} ({len(lrg_ran):,} randoms)")

print("\nPreview of LRG galaxy catalog:")
print(lrg_gal.head())
```

```python
# =====================================================
#  STEP 2 — Apply T-Web Classification (LRG Sample) ✅ FIXED
# =====================================================
import numpy as np, h5py, os
from scipy.ndimage import gaussian_filter
from tqdm import tqdm

DATA_DIR = os.path.expanduser("~/galaxy_evolution_project/LRG/output")
OUT_DIR  = DATA_DIR
os.makedirs(OUT_DIR, exist_ok=True)

# --- Load δ(x,y,z) field ---
delta = np.load(os.path.join(DATA_DIR, "delta_field_lrg.npy"))
N = delta.shape[0]
box_size = 640.0  # Mpc
dx = box_size / N
print(f"✅ Loaded density field: shape={delta.shape}, box={box_size} Mpc, dx={dx:.2f} Mpc")

# --- Parameters ---
R_smooth = 5.0   # Mpc
lambda_th = 0.0  # threshold

# --- 1. Smooth the field ---
sigma_vox = R_smooth / dx
delta_s = gaussian_filter(delta, sigma=sigma_vox)
print(f"✅ Smoothed δ-field with R_smooth = {R_smooth} Mpc ({sigma_vox:.2f} voxels)")

# --- 2. Compute gravitational potential via FFT ---
k = np.fft.fftfreq(N, d=dx) * 2 * np.pi
kx, ky, kz = np.meshgrid(k, k, k, indexing="ij")

k2 = kx**2 + ky**2 + kz**2
k2[0,0,0] = 1.0  # avoid div-by-zero
phi_k = -np.fft.fftn(delta_s) / k2
phi_k[0,0,0] = 0.0  # remove DC mode

# --- 3. Compute tidal tensor components ---
T = {}
for i, ki in enumerate([kx, ky, kz]):
    for j, kj in enumerate([kx, ky, kz]):
        Tij_k = (ki * kj) * phi_k
        T[(i, j)] = np.fft.ifftn(Tij_k).real

print("✅ Tidal tensor computed (9 components).")

# --- 4. Classify environments voxel-by-voxel ---
env_map = np.zeros((N, N, N), dtype=np.uint8)
labels = ["Void", "Sheet", "Filament", "Knot"]

print("🔄 Classifying voxels (this may take ~5–10 minutes)...")

for ix in tqdm(range(N), desc="T-Web Classification"):
    for iy in range(N):
        for iz in range(N):
            Tij = np.array([
                [T[(0,0)][ix,iy,iz], T[(0,1)][ix,iy,iz], T[(0,2)][ix,iy,iz]],
                [T[(1,0)][ix,iy,iz], T[(1,1)][ix,iy,iz], T[(1,2)][ix,iy,iz]],
                [T[(2,0)][ix,iy,iz], T[(2,1)][ix,iy,iz], T[(2,2)][ix,iy,iz]]
            ])
            eigvals = np.linalg.eigvalsh(Tij)
            env_map[ix,iy,iz] = np.sum(eigvals > lambda_th)

print("✅ Environment classification complete.")

# --- 5. Save map ---
out_h5 = os.path.join(OUT_DIR, "tweb_environment_map.h5")
with h5py.File(out_h5, "w") as f:
    f["env_map"] = env_map
print(f"💾 Saved environment map → {out_h5}")

# --- 6. Print fractions ---
counts = np.bincount(env_map.ravel(), minlength=4)
print("\n📊 Environment fractions:")
for i, n in enumerate(counts):
    frac = n / env_map.size * 100
    print(f"  {labels[i]:8s}: {frac:6.2f} %")
```

```python
# =====================================================
#  STEP 7 — Galaxy Counts per T-Web Environment (LRG)
# =====================================================

import pandas as pd
import numpy as np
import h5py, os
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors

# --- Paths ---
DATA_DIR = os.path.expanduser("~/galaxy_evolution_project/LRG/output")
CATALOG_FILE = os.path.expanduser("~/galaxy_evolution_project/LRG/lrg_galaxies_xyz.csv")
FIG_DIR = os.path.join(DATA_DIR, "figures")
os.makedirs(FIG_DIR, exist_ok=True)

# --- Load environment map ---
fname = os.path.join(DATA_DIR, "tweb_environment_map.h5")
with h5py.File(fname, "r") as f:
    env_map = f["env_map"][:]

N = env_map.shape[0]
box_size = 640.0  # [Mpc]
dx = box_size / N

# --- Load galaxy catalog ---
df = pd.read_csv(CATALOG_FILE)
print(f"✅ Loaded {len(df):,} LRG galaxies")

# --- Normalize galaxy positions into 0–640 Mpc cube ---
def rescale(arr, min_old, max_old, new_min=0, new_max=640):
    return (arr - min_old) / (max_old - min_old) * (new_max - new_min) + new_min

df["x_scaled"] = rescale(df["x"], df["x"].min(), df["x"].max())
df["y_scaled"] = rescale(df["y"], df["y"].min(), df["y"].max())
df["z_scaled"] = rescale(df["z"], df["z"].min(), df["z"].max())

# --- Map galaxies to voxel indices ---
ix = np.clip((df["x_scaled"] / dx).astype(int), 0, N - 1)
iy = np.clip((df["y_scaled"] / dx).astype(int), 0, N - 1)
iz = np.clip((df["z_scaled"] / dx).astype(int), 0, N - 1)

# --- Extract environment label for each galaxy ---
df["env_label"] = env_map[ix, iy, iz]

# --- Compute counts and fractions ---
labels = ["Void", "Sheet", "Filament", "Knot"]
colors = ["#1a1a1a", "#4ecdc4", "#0072b2", "#d73027"]

counts = df["env_label"].value_counts().reindex([0, 1, 2, 3]).fillna(0)
fractions = counts / counts.sum() * 100

# --- Print summary ---
print("\n📊 LRG Galaxy Distribution per T-Web Environment:")
for lab, n, f in zip(labels, counts, fractions):
    print(f"  {lab:8s}: {int(n):10,d} galaxies ({f:5.2f} %)")

# --- Plot ---
plt.style.use("seaborn-v0_8-white")
fig, ax = plt.subplots(figsize=(6.5, 4.5))

bars = ax.bar(labels, fractions, color=colors, edgecolor="black", linewidth=0.8, alpha=0.9)

# Add text labels on bars
for bar, frac in zip(bars, fractions):
    ax.text(bar.get_x() + bar.get_width() / 2, bar.get_height() + 0.4,
            f"{frac:.1f}%", ha='center', va='bottom', fontsize=11, fontweight='bold', color="black")

# --- Style ---
ax.set_title("Galaxy Counts per Web Environment — DESI LRGs (T-Web)", fontsize=14, weight="bold", pad=10)
ax.set_ylabel("Fraction of Galaxies [%]", fontsize=12)
ax.grid(axis='y', linestyle="--", alpha=0.3)
ax.set_facecolor("#f8f8f8")
for spine in ax.spines.values():
    spine.set_linewidth(1)

plt.tight_layout()

# --- Save ---
outfig = os.path.join(FIG_DIR, "tweb_galaxy_counts_LRG.png")
plt.savefig(outfig, dpi=450, bbox_inches="tight", facecolor="white")
plt.show()

print(f"💾 Saved → {outfig}")
```

```python
# =====================================================
#  STEP 9 — Evolution of LRG Populations vs Redshift (T-Web)
# =====================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import os
import h5py

# -----------------------------------------------------
# 1️⃣ Paths and Setup
# -----------------------------------------------------
DATA_DIR = os.path.expanduser("~/galaxy_evolution_project/LRG/output")
CATALOG_FILE = os.path.expanduser("~/galaxy_evolution_project/LRG/lrg_galaxies_xyz.csv")
FIG_DIR = os.path.join(DATA_DIR, "figures")
os.makedirs(FIG_DIR, exist_ok=True)

# -----------------------------------------------------
# 2️⃣ Load Galaxy Catalog and Environment Map
# -----------------------------------------------------
df = pd.read_csv(CATALOG_FILE)
print(f"✅ Loaded {len(df):,} LRGs with z range {df['Z'].min():.3f}–{df['Z'].max():.3f}")

# If env_label doesn't exist, create it from environment map
if "env_label" not in df.columns:
    print("🔄 Creating environment labels from T-Web map...")
    
    # Load environment map
    fname = os.path.join(DATA_DIR, "tweb_environment_map.h5")
    with h5py.File(fname, "r") as f:
        env_map = f["env_map"][:]
    
    N = env_map.shape[0]
    box_size = 640.0
    dx = box_size / N

    def rescale(arr, min_old, max_old, new_min=0, new_max=640):
        return (arr - min_old) / (max_old - min_old) * (new_max - new_min) + new_min

    df["x_scaled"] = rescale(df["x"], df["x"].min(), df["x"].max())
    df["y_scaled"] = rescale(df["y"], df["y"].min(), df["y"].max())
    df["z_scaled"] = rescale(df["z"], df["z"].min(), df["z"].max())

    ix = np.clip((df["x_scaled"] / dx).astype(int), 0, N - 1)
    iy = np.clip((df["y_scaled"] / dx).astype(int), 0, N - 1)
    iz = np.clip((df["z_scaled"] / dx).astype(int), 0, N - 1)

    df["env_label"] = env_map[ix, iy, iz]
    print("✅ Environment labels assigned to galaxies")
# -----------------------------------------------------
# 3️⃣ Define Redshift Bins
# -----------------------------------------------------
z_bins = np.arange(0.0, 1.2, 0.1)  # You can adjust Δz = 0.1
df["z_bin"] = pd.cut(df["Z"], bins=z_bins)

# -----------------------------------------------------
# 4️⃣ Compute Environment Fractions per Redshift Bin
# -----------------------------------------------------
labels = ["Void", "Sheet", "Filament", "Knot"]
env_ids = [0, 1, 2, 3]
results = []

for z_bin, group in df.groupby("z_bin"):
    total = len(group)
    if total == 0:
        continue
    counts = group["env_label"].value_counts().reindex(env_ids).fillna(0)
    fractions = counts / total * 100
    z_mid = z_bin.mid
    results.append([z_mid] + list(fractions))

df_evol = pd.DataFrame(results, columns=["z_mid"] + labels)
print("\n📊 Environment Fractions per Redshift Bin:")
print(df_evol.round(2))

# -----------------------------------------------------
# 5️⃣ Plot: Fraction of Galaxies vs Redshift
# -----------------------------------------------------
colors = ["#1a1a1a", "#4ecdc4", "#0072b2", "#d73027"]  # same color scheme

plt.style.use("seaborn-v0_8-whitegrid")
fig, ax = plt.subplots(figsize=(7.5, 5))

for i, lab in enumerate(labels):
    ax.plot(df_evol["z_mid"], df_evol[lab], marker="o", lw=2.2,
            color=colors[i], label=lab)

# Aesthetics
ax.set_xlabel("Redshift z", fontsize=12)
ax.set_ylabel("Fraction of Galaxies [%]", fontsize=12)
ax.set_title("Evolution of LRG Populations by Cosmic Web Environment (T-Web)", 
             fontsize=14, weight="bold", pad=10)
ax.legend(title="Environment", frameon=True, fontsize=10)
ax.grid(alpha=0.3)
ax.set_xlim(df_evol["z_mid"].min() - 0.05, df_evol["z_mid"].max() + 0.05)
ax.set_ylim(0, df_evol[labels].max().max() * 1.2)

plt.tight_layout()

# -----------------------------------------------------
# 6️⃣ Save Figure
# -----------------------------------------------------
outfig = os.path.join(FIG_DIR, "tweb_environment_fractions_vs_redshift_LRG.png")
plt.savefig(outfig, dpi=450, bbox_inches="tight", facecolor="white")
plt.show()

print(f"💾 Saved → {outfig}")
```

```python
# =====================================================
#  STEP 10 — T-Web Redshift Evolution with Error Bars
# =====================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import os
import h5py
from scipy import stats

# -----------------------------------------------------
# Paths
# -----------------------------------------------------
DATA_DIR = os.path.expanduser("~/galaxy_evolution_project/LRG/output")
CATALOG_FILE = os.path.expanduser("~/galaxy_evolution_project/LRG/lrg_galaxies_xyz.csv")
FIG_DIR = os.path.join(DATA_DIR, "figures")
os.makedirs(FIG_DIR, exist_ok=True)

# -----------------------------------------------------
# Load Galaxy Catalog
# -----------------------------------------------------
df = pd.read_csv(CATALOG_FILE)
print(f"✅ Loaded {len(df):,} LRGs")

# Assign environment labels if missing
if "env_label" not in df.columns:
    print("⚙️  Assigning env_label automatically...")
    fname = os.path.join(DATA_DIR, "tweb_environment_map.h5")
    with h5py.File(fname, "r") as f:
        env_map = f["env_map"][:]
    
    N = env_map.shape[0]
    box_size = 640.0
    dx = box_size / N
    
    def rescale(arr, min_old, max_old, new_min=0, new_max=640):
        return (arr - min_old) / (max_old - min_old) * (new_max - new_min) + new_min
    
    df["x_scaled"] = rescale(df["x"], df["x"].min(), df["x"].max())
    df["y_scaled"] = rescale(df["y"], df["y"].min(), df["y"].max())
    df["z_scaled"] = rescale(df["z"], df["z"].min(), df["z"].max())
    
    ix = np.clip((df["x_scaled"] / dx).astype(int), 0, N - 1)
    iy = np.clip((df["y_scaled"] / dx).astype(int), 0, N - 1)
    iz = np.clip((df["z_scaled"] / dx).astype(int), 0, N - 1)
    
    df["env_label"] = env_map[ix, iy, iz]
    print("✅ Environment labels assigned")

# -----------------------------------------------------
# Limit to realistic DESI LRG redshifts and bin
# -----------------------------------------------------
df = df[df["Z"] < 1.2].copy()
print(f"📊 Using {len(df):,} galaxies with z < 1.2")

z_bins = np.linspace(0.0, 1.1, 12)  # More bins for better resolution
df["z_bin"] = pd.cut(df["Z"], bins=z_bins)

# -----------------------------------------------------
# Compute fractions with proper Poisson errors
# -----------------------------------------------------
labels = ["Void", "Sheet", "Filament", "Knot"]
env_ids = [0, 1, 2, 3]
results = []

for z_bin, group in df.groupby("z_bin", observed=False):  # Fix the warning
    total = len(group)
    if total < 10:  # Skip bins with too few galaxies
        continue
        
    counts = group["env_label"].value_counts().reindex(env_ids).fillna(0)
    fractions = counts / total * 100
    
    # Calculate proper Poisson errors using Wilson score interval
    errors = []
    for count in counts:
        if count > 0 and total > 0:
            # Wilson score interval (68% confidence)
            p = count / total
            z = 1.0  # 1 sigma for 68% CI
            denominator = 1 + z**2 / total
            center = (p + z**2 / (2 * total)) / denominator
            margin = z * np.sqrt(p * (1 - p) / total + z**2 / (4 * total**2)) / denominator
            
            ci_low = center - margin
            ci_high = center + margin
            
            # Convert to percentage and get error
            error_low = fractions[counts.index[counts == count].values[0]] - ci_low * 100
            error_high = ci_high * 100 - fractions[counts.index[counts == count].values[0]]
            error = max(error_low, error_high)
        else:
            error = 0.1  # Small minimum error for zero counts
        errors.append(error)
    
    z_mid = z_bin.mid
    results.append([z_mid, total] + list(fractions) + errors)

cols = ["z_mid", "N_galaxies"] + labels + [f"{lab}_err" for lab in labels]
df_evol = pd.DataFrame(results, columns=cols)

print("\n📊 Environment Fractions with Errors:")
print(df_evol.round(2))

# -----------------------------------------------------
# Plot with proper error bars and shaded bands
# -----------------------------------------------------
colors = ["#1a1a1a", "#4ecdc4", "#0072b2", "#d73027"]
markers = ["o", "s", "D", "^"]

plt.style.use("seaborn-v0_8-whitegrid")
fig, ax = plt.subplots(figsize=(10, 6))

for i, lab in enumerate(labels):
    z_values = df_evol["z_mid"]
    frac_values = df_evol[lab]
    err_values = df_evol[f"{lab}_err"]
    
    # Main line with markers
    ax.plot(z_values, frac_values, color=colors[i], linewidth=2.5, 
            marker=markers[i], markersize=6, markeredgecolor='white', 
            markeredgewidth=1, label=lab, zorder=5)
    
    # Error bars
    ax.errorbar(z_values, frac_values, yerr=err_values, fmt='none',
                ecolor=colors[i], elinewidth=1.5, capsize=4, capthick=1.5,
                alpha=0.7, zorder=4)
    
    # Shaded error band
    ax.fill_between(z_values, frac_values - err_values, frac_values + err_values,
                    color=colors[i], alpha=0.15, zorder=3)

# -----------------------------------------------------
# Enhanced styling
# -----------------------------------------------------
ax.set_xlabel("Redshift z", fontsize=14, fontweight='bold')
ax.set_ylabel("Fraction of Galaxies [%]", fontsize=14, fontweight='bold')
ax.set_title("Evolution of LRG Populations by Cosmic Web Environment (T-Web)",
             fontsize=16, fontweight='bold', pad=20)

ax.legend(title="Environment", title_fontsize=12, fontsize=11, 
          frameon=True, fancybox=True, shadow=True, loc='best')

ax.grid(True, alpha=0.3, linestyle='--', linewidth=0.8)
ax.set_xlim(-0.05, 1.15)
ax.set_ylim(0, max(df_evol[labels].max().max() * 1.15, 40))



# -----------------------------------------------------
# Save high-quality figure
# -----------------------------------------------------
outfig = os.path.join(FIG_DIR, "tweb_environment_evolution_with_errors.png")
plt.savefig(outfig, dpi=450, bbox_inches='tight', facecolor='white', 
            transparent=False, format='png')
plt.show()

print(f"💾 Saved → {outfig}")
print(f"📈 Plot shows evolution across {len(df_evol)} redshift bins")
```

```python
# ==============================================================
#  Cell 11 — Systematic T-Web parameter scan (R_smooth, d_border)
# ==============================================================

import numpy as np, h5py, os, pandas as pd
from scipy.ndimage import gaussian_filter
from tqdm import tqdm

DATA_DIR = os.path.expanduser("~/galaxy_evolution_project/LRG/output")
FIG_DIR  = os.path.join(DATA_DIR, "../figures")
os.makedirs(FIG_DIR, exist_ok=True)

# --------------------------------------------------------------
# Load density contrast field δ(x)
# --------------------------------------------------------------
dens_file = os.path.join(DATA_DIR, "density_field_lrg.h5")
with h5py.File(dens_file, "r") as f:
    delta = f["delta"][:]
print(f"✅ Loaded δ-field: shape = {delta.shape}")

N = delta.shape[0]

# --------------------------------------------------------------
# Parameter grid to explore
# --------------------------------------------------------------
R_smooth_values = [3, 5, 10]   # Mpc/h
d_border_values = [0, 10, 20]  # Mpc (approx.)

results = []

# --------------------------------------------------------------
# Helper: compute simple tidal eigenvalue count (toy T-Web)
# --------------------------------------------------------------
def classify_tweb_field(field, R_smooth, d_border_vox, lam_th=0.2):
    # Smooth field
    field_smoothed = gaussian_filter(field, sigma=R_smooth)
    # Compute Hessian by finite differences
    gx, gy, gz = np.gradient(field_smoothed)
    Hxx, Hxy, Hxz = np.gradient(gx)
    Hyx, Hyy, Hyz = np.gradient(gy)
    Hzx, Hzy, Hzz = np.gradient(gz)
    # Symmetrize
    T11, T12, T13 = Hxx, 0.5*(Hxy+Hyx), 0.5*(Hxz+Hzx)
    T22, T23 = Hyy, 0.5*(Hyz+Hzy)
    T33 = Hzz
    # Eigen decomposition voxel-by-voxel
    env_map = np.zeros_like(field, dtype=np.uint8)
    for i in tqdm(range(N), desc=f"λ-count R={R_smooth}", leave=False):
        for j in range(N):
            for k in range(N):
                T = np.array([[T11[i,j,k], T12[i,j,k], T13[i,j,k]],
                              [T12[i,j,k], T22[i,j,k], T23[i,j,k]],
                              [T13[i,j,k], T23[i,j,k], T33[i,j,k]]])
                eigvals = np.linalg.eigvalsh(T)
                n_collapse = np.sum(eigvals > lam_th)
                env_map[i,j,k] = n_collapse
    # Apply border mask
    if d_border_vox > 0:
        env_map[:d_border_vox,:,:] = 0
        env_map[-d_border_vox:,:,:] = 0
        env_map[:,:d_border_vox,:] = 0
        env_map[:,-d_border_vox:,:] = 0
        env_map[:,:,:d_border_vox] = 0
        env_map[:,:,-d_border_vox:] = 0
    return env_map

# --------------------------------------------------------------
# Loop over parameter combinations
# --------------------------------------------------------------
for R_smooth in R_smooth_values:
    for d_border in d_border_values:
        print(f"\n🔄 Running T-Web: R_smooth={R_smooth}  d_border={d_border}")
        env_map = classify_tweb_field(delta, R_smooth, int(d_border/5))  # rough voxel conversion

        # Save map
        outname = f"tweb_environment_R{R_smooth}_d{d_border}.h5"
        with h5py.File(os.path.join(DATA_DIR, outname), "w") as f:
            f.create_dataset("env_map", data=env_map, compression="gzip")
        print(f"💾 Saved → {outname}")

        # Compute fractions
        counts = np.bincount(env_map.ravel(), minlength=4)
        fracs = counts / counts.sum()
        results.append({"R_smooth": R_smooth, "d_border": d_border,
                        "Void": fracs[0], "Sheet": fracs[1],
                        "Filament": fracs[2], "Knot": fracs[3]})
        print(f"Fractions: Void={fracs[0]*100:.1f}%  Filament={fracs[2]*100:.1f}%  Knot={fracs[3]*100:.1f}%")

# --------------------------------------------------------------
# Save summary table
# --------------------------------------------------------------
df_summary = pd.DataFrame(results)
outcsv = os.path.join(DATA_DIR, "tweb_scale_dependence_summary.csv")
df_summary.to_csv(outcsv, index=False)
print(f"\n📊 Saved summary table → {outcsv}")
print(df_summary)
```

```python
# ==============================================================
#  Cell 11 — Physically accurate T-Web multi-scale scan (FFT + vectorized)
# ==============================================================

import numpy as np, h5py, pandas as pd, os
from scipy.ndimage import gaussian_filter
from scipy.fft import fftn, ifftn, fftfreq
from tqdm import tqdm

DATA_DIR = os.path.expanduser("~/galaxy_evolution_project/LRG/output")
FIG_DIR  = os.path.join(DATA_DIR, "../figures")
os.makedirs(FIG_DIR, exist_ok=True)

# --------------------------------------------------------------
# 1. Load δ-field
# --------------------------------------------------------------
dens_file = os.path.join(DATA_DIR, "density_field_lrg.h5")
with h5py.File(dens_file, "r") as f:
    delta = f["delta"][:]
print(f"✅ Loaded δ-field: shape = {delta.shape}")

N = delta.shape[0]
delta = delta - np.mean(delta)  # ensure mean-zero

# --------------------------------------------------------------
# 2. Define parameter grid
# --------------------------------------------------------------
R_smooth_values = [1, 3, 5, 10]   # Mpc/h
d_border_values = [0, 10, 20]     # Mpc (approx.)
lam_th = 0.05                     # eigenvalue threshold
results = []

# --------------------------------------------------------------
# 3. Precompute k-grids for FFT derivatives
# --------------------------------------------------------------
kx = ky = kz = fftfreq(N) * 2 * np.pi
kx, ky, kz = np.meshgrid(kx, ky, kz, indexing="ij")
k2 = kx**2 + ky**2 + kz**2
k2[0, 0, 0] = 1.0  # avoid divide-by-zero

# --------------------------------------------------------------
# Helper: classify via FFT tidal tensor + vectorized eigvals
# --------------------------------------------------------------
def classify_tweb_fft(delta, R_smooth, d_border_vox, lam_th=0.05):
    # 1. Smooth δ field
    delta_smooth = gaussian_filter(delta, sigma=R_smooth)

    # 2. Solve Poisson equation: ∇²Φ = δ → Φ(k) = -δ(k)/k²
    delta_k = fftn(delta_smooth)
    phi_k = -delta_k / k2

    # 3. Compute Hessian (tidal tensor) components via FFT multiplications
    Txx = np.real(ifftn((kx*kx) * phi_k))
    Tyy = np.real(ifftn((ky*ky) * phi_k))
    Tzz = np.real(ifftn((kz*kz) * phi_k))
    Txy = np.real(ifftn((kx*ky) * phi_k))
    Txz = np.real(ifftn((kx*kz) * phi_k))
    Tyz = np.real(ifftn((ky*kz) * phi_k))

    # 4. Flatten grids and compute eigvals vectorized
    n_vox = N**3
    tensors = np.zeros((n_vox, 3, 3), dtype=np.float32)
    tensors[:,0,0] = Txx.ravel()
    tensors[:,1,1] = Tyy.ravel()
    tensors[:,2,2] = Tzz.ravel()
    tensors[:,0,1] = tensors[:,1,0] = Txy.ravel()
    tensors[:,0,2] = tensors[:,2,0] = Txz.ravel()
    tensors[:,1,2] = tensors[:,2,1] = Tyz.ravel()

    eigvals = np.linalg.eigvalsh(tensors)  # shape (N^3, 3)
    n_collapse = np.sum(eigvals > lam_th, axis=1).astype(np.uint8)
    env_map = n_collapse.reshape((N, N, N))

    # 5. Mask borders if needed
    if d_border_vox > 0:
        env_map[:d_border_vox,:,:] = 0
        env_map[-d_border_vox:,:,:] = 0
        env_map[:,:d_border_vox,:] = 0
        env_map[:,-d_border_vox:,:] = 0
        env_map[:,:,:d_border_vox] = 0
        env_map[:,:,-d_border_vox:] = 0

    return env_map

# --------------------------------------------------------------
# 4. Run parameter grid
# --------------------------------------------------------------
for R_smooth in R_smooth_values:
    for d_border in d_border_values:
        print(f"\n🔄 Running T-Web: R_smooth={R_smooth}  d_border={d_border}")

        # convert border size (Mpc) to voxel count roughly
        d_border_vox = int((d_border / 4000.0) * N)  # assuming ~4 Gpc box
        env_map = classify_tweb_fft(delta, R_smooth, d_border_vox, lam_th)

        # Save HDF5 map
        outname = f"tweb_environment_R{R_smooth}_d{d_border}.h5"
        with h5py.File(os.path.join(DATA_DIR, outname), "w") as f:
            f.create_dataset("env_map", data=env_map, compression="gzip")
        print(f"💾 Saved → {outname}")

        # Compute fractions
        counts = np.bincount(env_map.ravel(), minlength=4)
        fracs = counts / counts.sum()
        results.append({
            "R_smooth": R_smooth, "d_border": d_border,
            "Void": fracs[0], "Sheet": fracs[1],
            "Filament": fracs[2], "Knot": fracs[3]
        })
        print(f"Fractions: Void={fracs[0]*100:.1f}%  Sheet={fracs[1]*100:.1f}%  "
              f"Filament={fracs[2]*100:.1f}%  Knot={fracs[3]*100:.1f}%")

# --------------------------------------------------------------
# 5. Save summary CSV
# --------------------------------------------------------------
df_summary = pd.DataFrame(results)
outcsv = os.path.join(DATA_DIR, "tweb_scale_dependence_summary.csv")
df_summary.to_csv(outcsv, index=False)
print(f"\n📊 Saved summary table → {outcsv}")
print(df_summary)
```

```python
# ==============================================================
#  ASTRA-lite Cosmic Web Classification (Self-Contained)
#  --------------------------------------------------------------
#  Computes: smoothed density field → tidal tensor → eigenvalues
#  → classification → volume fractions → HDF5 output
# ==============================================================

import numpy as np
import h5py, os
from scipy.ndimage import gaussian_filter
from numpy.fft import fftn, ifftn

# --------------------------------------------------------------
# User Paths
# --------------------------------------------------------------
DATA_DIR = os.path.expanduser("~/galaxy_evolution_project/LRG/output")
dens_file = os.path.join(DATA_DIR, "density_field_lrg.h5")

# --------------------------------------------------------------
# 1️⃣ Load Density Field
# --------------------------------------------------------------
with h5py.File(dens_file, "r") as f:
    delta = f["delta"][:]

print(f"✅ Loaded DESI LRG δ-field: shape={delta.shape}, mean={np.mean(delta):.3e}")

# --------------------------------------------------------------
# 2️⃣ Define ASTRA-lite helper functions
# --------------------------------------------------------------

def smooth_field(field, R_smooth, box_size, grid_size):
    """Apply Gaussian smoothing in real space."""
    dx = box_size / grid_size
    sigma_vox = R_smooth / dx
    return gaussian_filter(field, sigma=sigma_vox)

def calculate_tidal_tensor(delta_field):
    """Compute the tidal tensor T_ij = ∂²Φ/∂x_i∂x_j via FFT Poisson solver."""
    N = delta_field.shape[0]
    kfreq = np.fft.fftfreq(N) * 2 * np.pi
    kx, ky, kz = np.meshgrid(kfreq, kfreq, kfreq, indexing="ij")

    k2 = kx**2 + ky**2 + kz**2
    k2[0,0,0] = 1.0  # avoid div-by-zero

    # Poisson equation in Fourier space: Φ(k) = -δ(k)/k²
    phi_k = -fftn(delta_field) / k2
    phi_k[0,0,0] = 0.0

    # Compute tidal tensor components
    T = np.zeros((3,3,N,N,N), dtype=np.float32)
    for i, ki in enumerate([kx, ky, kz]):
        for j, kj in enumerate([kx, ky, kz]):
            T[i,j] = np.real(ifftn((ki * kj) * phi_k))
    return T

def calculate_eigenvalues_and_vectors(T, N):
    """Compute eigenvalues and eigenvectors of tidal tensor at each voxel."""
    eigvals = np.zeros((N,N,N,3), dtype=np.float32)
    eigvecs = np.zeros((N,N,N,3,3), dtype=np.float32)
    for i in range(N):
        for j in range(N):
            Tij = np.array([
                [T[0,0,i,j,:], T[0,1,i,j,:], T[0,2,i,j,:]],
                [T[1,0,i,j,:], T[1,1,i,j,:], T[1,2,i,j,:]],
                [T[2,0,i,j,:], T[2,1,i,j,:], T[2,2,i,j,:]]
            ])
            # Each voxel along z-axis
            for k in range(N):
                mat = np.array([[T[0,0,i,j,k], T[0,1,i,j,k], T[0,2,i,j,k]],
                                [T[1,0,i,j,k], T[1,1,i,j,k], T[1,2,i,j,k]],
                                [T[2,0,i,j,k], T[2,1,i,j,k], T[2,2,i,j,k]]])
                vals, vecs = np.linalg.eigh(mat)
                eigvals[i,j,k,:] = vals
                eigvecs[i,j,k,:,:] = vecs
    return eigvals, eigvecs

def classify_structure(eigvals, lambda_th=0.05):
    """Classify each voxel: 0=Void, 1=Sheet, 2=Filament, 3=Knot."""
    env_map = np.sum(eigvals > lambda_th, axis=-1)
    return env_map.astype(np.uint8)

def calculate_volume_fractions(env_map):
    """Compute fractional volume for each web type."""
    counts = np.bincount(env_map.ravel(), minlength=4)
    fractions = counts / env_map.size
    return fractions

# --------------------------------------------------------------
# 3️⃣ Set Parameters
# --------------------------------------------------------------
R_smooth = 5.0    # Mpc/h
BOX_SIZE = 4000.0 # Mpc/h
GRID_SIZE = delta.shape[0]
LAMBDA_TH = 0.05  # Threshold

print(f"\n🔧 Parameters: R_smooth={R_smooth} Mpc/h | Box={BOX_SIZE} Mpc/h | Grid={GRID_SIZE}³")

# --------------------------------------------------------------
# 4️⃣ Run ASTRA-lite pipeline
# --------------------------------------------------------------
smoothed = smooth_field(delta, R_smooth, BOX_SIZE, GRID_SIZE)
print("✅ Smoothed field computed")

tidal_tensor = calculate_tidal_tensor(smoothed)
print("✅ Tidal tensor computed")

eigvals, eigvecs = calculate_eigenvalues_and_vectors(tidal_tensor, GRID_SIZE)
print("✅ Eigenvalues and eigenvectors computed")

classification = classify_structure(eigvals, LAMBDA_TH)
print("✅ Environment classification complete")

fractions = calculate_volume_fractions(classification)
labels = ["Void", "Sheet", "Filament", "Knot"]

print("\n📊 ASTRA-lite Volume Fractions:")
for label, frac in zip(labels, fractions):
    print(f"  {label:8s}: {frac*100:6.2f}%")

# --------------------------------------------------------------
# 5️⃣ Save Environment Map
# --------------------------------------------------------------
outname = os.path.join(DATA_DIR, "astra_environment_map.h5")
with h5py.File(outname, "w") as f:
    f.create_dataset("env_map", data=classification, compression="gzip")

print(f"\n💾 Saved ASTRA environment map → {outname}")
```

```python
# =====================================================
#  STEP 10 — T-Web Redshift Evolution with Error Bars
# =====================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import os
import h5py
from scipy import stats

# -----------------------------------------------------
# Paths
# -----------------------------------------------------
DATA_DIR = os.path.expanduser("~/galaxy_evolution_project/LRG/output")
CATALOG_FILE = os.path.expanduser("~/galaxy_evolution_project/LRG/lrg_galaxies_xyz.csv")
FIG_DIR = os.path.join(DATA_DIR, "figures")
os.makedirs(FIG_DIR, exist_ok=True)

# -----------------------------------------------------
# Load Galaxy Catalog
# -----------------------------------------------------
df = pd.read_csv(CATALOG_FILE)
print(f"✅ Loaded {len(df):,} LRGs")

# Assign environment labels if missing
if "env_label" not in df.columns:
    print("⚙️  Assigning env_label automatically...")
    fname = os.path.join(DATA_DIR, "astra_environment_map.h5")
    with h5py.File(fname, "r") as f:
        env_map = f["env_map"][:]
    
    N = env_map.shape[0]
    box_size = 640.0
    dx = box_size / N
    
    def rescale(arr, min_old, max_old, new_min=0, new_max=640):
        return (arr - min_old) / (max_old - min_old) * (new_max - new_min) + new_min
    
    df["x_scaled"] = rescale(df["x"], df["x"].min(), df["x"].max())
    df["y_scaled"] = rescale(df["y"], df["y"].min(), df["y"].max())
    df["z_scaled"] = rescale(df["z"], df["z"].min(), df["z"].max())
    
    ix = np.clip((df["x_scaled"] / dx).astype(int), 0, N - 1)
    iy = np.clip((df["y_scaled"] / dx).astype(int), 0, N - 1)
    iz = np.clip((df["z_scaled"] / dx).astype(int), 0, N - 1)
    
    df["env_label"] = env_map[ix, iy, iz]
    print("✅ Environment labels assigned")

# -----------------------------------------------------
# Limit to realistic DESI LRG redshifts and bin
# -----------------------------------------------------
df = df[df["Z"] < 1.2].copy()
print(f"📊 Using {len(df):,} galaxies with z < 1.2")

z_bins = np.linspace(0.0, 1.1, 12)  # More bins for better resolution
df["z_bin"] = pd.cut(df["Z"], bins=z_bins)

# -----------------------------------------------------
# Compute fractions with proper Poisson errors
# -----------------------------------------------------
labels = ["Void", "Sheet", "Filament", "Knot"]
env_ids = [0, 1, 2, 3]
results = []

for z_bin, group in df.groupby("z_bin", observed=False):  # Fix the warning
    total = len(group)
    if total < 10:  # Skip bins with too few galaxies
        continue
        
    counts = group["env_label"].value_counts().reindex(env_ids).fillna(0)
    fractions = counts / total * 100
    
    # Calculate proper Poisson errors using Wilson score interval
    errors = []
    for count in counts:
        if count > 0 and total > 0:
            # Wilson score interval (68% confidence)
            p = count / total
            z = 1.0  # 1 sigma for 68% CI
            denominator = 1 + z**2 / total
            center = (p + z**2 / (2 * total)) / denominator
            margin = z * np.sqrt(p * (1 - p) / total + z**2 / (4 * total**2)) / denominator
            
            ci_low = center - margin
            ci_high = center + margin
            
            # Convert to percentage and get error
            error_low = fractions[counts.index[counts == count].values[0]] - ci_low * 100
            error_high = ci_high * 100 - fractions[counts.index[counts == count].values[0]]
            error = max(error_low, error_high)
        else:
            error = 0.1  # Small minimum error for zero counts
        errors.append(error)
    
    z_mid = z_bin.mid
    results.append([z_mid, total] + list(fractions) + errors)

cols = ["z_mid", "N_galaxies"] + labels + [f"{lab}_err" for lab in labels]
df_evol = pd.DataFrame(results, columns=cols)

print("\n📊 Environment Fractions with Errors:")
print(df_evol.round(2))

# -----------------------------------------------------
# Plot with proper error bars and shaded bands
# -----------------------------------------------------
colors = ["#1a1a1a", "#4ecdc4", "#0072b2", "#d73027"]
markers = ["o", "s", "D", "^"]

plt.style.use("seaborn-v0_8-whitegrid")
fig, ax = plt.subplots(figsize=(10, 6))

for i, lab in enumerate(labels):
    z_values = df_evol["z_mid"]
    frac_values = df_evol[lab]
    err_values = df_evol[f"{lab}_err"]
    
    # Main line with markers
    ax.plot(z_values, frac_values, color=colors[i], linewidth=2.5, 
            marker=markers[i], markersize=6, markeredgecolor='white', 
            markeredgewidth=1, label=lab, zorder=5)
    
    # Error bars
    ax.errorbar(z_values, frac_values, yerr=err_values, fmt='none',
                ecolor=colors[i], elinewidth=1.5, capsize=4, capthick=1.5,
                alpha=0.7, zorder=4)
    
    # Shaded error band
    ax.fill_between(z_values, frac_values - err_values, frac_values + err_values,
                    color=colors[i], alpha=0.15, zorder=3)

# -----------------------------------------------------
# Enhanced styling
# -----------------------------------------------------
ax.set_xlabel("Redshift z", fontsize=14, fontweight='bold')
ax.set_ylabel("Fraction of Galaxies [%]", fontsize=14, fontweight='bold')
ax.set_title("Evolution of LRG Populations by Cosmic Web Environment (ASTRA)",
             fontsize=16, fontweight='bold', pad=20)

ax.legend(title="Environment", title_fontsize=12, fontsize=11, 
          frameon=True, fancybox=True, shadow=True, loc='best')

ax.grid(True, alpha=0.3, linestyle='--', linewidth=0.8)
ax.set_xlim(-0.05, 1.15)
ax.set_ylim(0, max(df_evol[labels].max().max() * 1.15, 40))



# -----------------------------------------------------
# Save high-quality figure
# -----------------------------------------------------
outfig = os.path.join(FIG_DIR, "ASTRA_environment_evolution_with_errors.png")
plt.savefig(outfig, dpi=450, bbox_inches='tight', facecolor='white', 
            transparent=False, format='png')
plt.show()

print(f"💾 Saved → {outfig}")
print(f"📈 Plot shows evolution across {len(df_evol)} redshift bins")
```

```python
# ==============================================================
#  Systematic ASTRA Parameter Scan: (R_smooth, d_border)
# ==============================================================
import numpy as np, os, h5py, pandas as pd
from scipy.ndimage import gaussian_filter
from tqdm import tqdm
import matplotlib.pyplot as plt

# ==============================================================
# 1️⃣ Load density field
# ==============================================================
DATA_DIR = os.path.expanduser("~/galaxy_evolution_project/LRG/output")
dens_file = os.path.join(DATA_DIR, "density_field_lrg.h5")

with h5py.File(dens_file, "r") as f:
    delta = f["delta"][:]
print(f"✅ Loaded density field: shape={delta.shape}, mean δ={np.mean(delta):.3e}")

N = delta.shape[0]
BOX_SIZE = 640.0  # Mpc/h
dx = BOX_SIZE / N

# ==============================================================
# 2️⃣ Helper functions for ASTRA
# ==============================================================
def smooth_field(field, R_smooth, box_size, grid_size):
    """Gaussian smoothing in 3D"""
    sigma_vox = R_smooth / (box_size / grid_size)
    return gaussian_filter(field, sigma=sigma_vox)

def compute_tidal_tensor(delta_field):
    """Compute Hessian (tidal tensor) of potential ∇∇Φ ∝ ∇∇∇⁻²δ"""
    k = np.fft.fftfreq(delta_field.shape[0]) * 2 * np.pi
    kx, ky, kz = np.meshgrid(k, k, k, indexing="ij")
    k2 = kx**2 + ky**2 + kz**2
    k2[0, 0, 0] = 1.0  # avoid divide by zero

    delta_k = np.fft.fftn(delta_field)
    phi_k = -delta_k / k2
    phi_k[0, 0, 0] = 0.0

    T = {}
    for i, ki in enumerate([kx, ky, kz]):
        for j, kj in enumerate([kx, ky, kz]):
            T[(i, j)] = np.fft.ifftn(ki * kj * phi_k).real
    return T

def classify_ast(T, lam_th=0.05):
    """Classify voxels based on eigenvalue thresholds"""
    N = list(T.values())[0].shape[0]
    env = np.zeros((N, N, N), dtype=np.uint8)
    for ix in range(N):
        for iy in range(N):
            Tij = np.array([
                [T[(0, 0)][ix, iy, :], T[(0, 1)][ix, iy, :], T[(0, 2)][ix, iy, :]],
                [T[(1, 0)][ix, iy, :], T[(1, 1)][ix, iy, :], T[(1, 2)][ix, iy, :]],
                [T[(2, 0)][ix, iy, :], T[(2, 1)][ix, iy, :], T[(2, 2)][ix, iy, :]]
            ])
            eigvals = np.linalg.eigvalsh(Tij)
            env[ix, iy, :] = np.sum(eigvals > lam_th)
    return env

def volume_fractions(env_map):
    """Return fraction of each environment type"""
    counts = np.bincount(env_map.ravel(), minlength=4)
    return counts / np.sum(counts)

# ==============================================================
# 3️⃣ Parameter ranges
# ==============================================================
R_smooth_values = [3, 5, 8, 10]     # in Mpc/h
d_border_values = [5, 10, 15]       # in Mpc/h

results = []

def classify_ast(T, lam_th=0.05):
    """
    Classify voxels based on eigenvalues of the tidal tensor.
    For each voxel, count how many eigenvalues > lambda_th.
    """
    N = list(T.values())[0].shape[0]
    env = np.zeros((N, N, N), dtype=np.uint8)

    for ix in range(N):
        for iy in range(N):
            for iz in range(N):
                Tij = np.array([
                    [T[(0, 0)][ix, iy, iz], T[(0, 1)][ix, iy, iz], T[(0, 2)][ix, iy, iz]],
                    [T[(1, 0)][ix, iy, iz], T[(1, 1)][ix, iy, iz], T[(1, 2)][ix, iy, iz]],
                    [T[(2, 0)][ix, iy, iz], T[(2, 1)][ix, iy, iz], T[(2, 2)][ix, iy, iz]]
                ])
                eigvals = np.linalg.eigvalsh(Tij)
                env[ix, iy, iz] = np.sum(eigvals > lam_th)
    return env


# ==============================================================
# 4️⃣ Main Scan Loop
# ==============================================================
for R_smooth in R_smooth_values:
    for d_border in d_border_values:
        print(f"\n🌀 Running ASTRA with R_smooth={R_smooth} Mpc/h, d_border={d_border} Mpc/h")

        # --- Smooth ---
        smoothed = smooth_field(delta, R_smooth, BOX_SIZE, N)

        # --- Compute tidal tensor ---
        T = compute_tidal_tensor(smoothed)

        # --- Classify ---
        env_map = classify_ast(T)

        # --- Mask border voxels ---
        border_vox = int(d_border / dx)
        env_central = env_map[border_vox:-border_vox, border_vox:-border_vox, border_vox:-border_vox]

        # --- Compute fractions ---
        fractions = volume_fractions(env_central)

        results.append({
            "R_smooth": R_smooth,
            "d_border": d_border,
            "Void": fractions[0],
            "Sheet": fractions[1],
            "Filament": fractions[2],
            "Knot": fractions[3]
        })

        print(f"✅ Volume fractions: Void={fractions[0]*100:.2f}%, Sheet={fractions[1]*100:.2f}%, "
              f"Filament={fractions[2]*100:.2f}%, Knot={fractions[3]*100:.2f}%")

# ==============================================================
# 5️⃣ Save results
# ==============================================================
df = pd.DataFrame(results)
out_csv = os.path.join(DATA_DIR, "astra_parameter_scan.csv")
df.to_csv(out_csv, index=False)
print(f"\n💾 Saved scan results → {out_csv}")

# ==============================================================
# 6️⃣ Visualization with T-Web Matching Colors
# ==============================================================

tweb_colors = {
    "Void":     "#1b1b1b",  # charcoal/black
    "Sheet":    "#4daf4a",  # green
    "Filament": "#377eb8",  # blue
    "Knot":     "#e41a1c"   # red
}

plt.figure(figsize=(7, 5))

for env in ["Void", "Sheet", "Filament", "Knot"]:
    plt.plot(
        df["R_smooth"], df[env] * 100,
        "o-", linewidth=2.0, markersize=7,
        label=env, color=tweb_colors[env]
    )

plt.xlabel("R_smooth [Mpc/h]", fontsize=12)
plt.ylabel("Volume Fraction [%]", fontsize=12)
plt.title("ASTRA Volume Fractions vs Smoothing Scale", fontsize=14, weight="bold")
plt.grid(alpha=0.25)

plt.legend(title="Environment", fontsize=11)
plt.tight_layout()

outfig = os.path.join(DATA_DIR, "astra_scan_summary_matched_colors.png")
plt.savefig(outfig, dpi=450, bbox_inches="tight")
plt.show()

print(f"💾 Saved matched-color figure → {outfig}")
```

```python
# =====================================================
#  CONFUSION MATRIX — T-Web vs ASTRA (LRG sample)
# =====================================================

import pandas as pd
import numpy as np
import h5py, os
import matplotlib.pyplot as plt
import seaborn as sns

# ---------------------------
# Paths
# ---------------------------
BASE = os.path.expanduser("~/galaxy_evolution_project/LRG/output")
FIG_DIR = os.path.join(BASE, "figures")
os.makedirs(FIG_DIR, exist_ok=True)

CATALOG = os.path.expanduser("~/galaxy_evolution_project/LRG/lrg_galaxies_xyz.csv")
TWEB_MAP  = os.path.join(BASE, "tweb_environment_map.h5")
ASTRA_MAP = os.path.join(BASE, "astra_environment_map.h5")

# ---------------------------
# Load environment maps
# ---------------------------
with h5py.File(TWEB_MAP, "r") as f:
    env_tweb = f["env_map"][:]

with h5py.File(ASTRA_MAP, "r") as f:
    env_astra = f["env_map"][:]

N = env_tweb.shape[0]
BOX = 640.0
dx  = BOX / N

labels = ["Void", "Sheet", "Filament", "Knot"]

print("Loaded T-Web:", env_tweb.shape)
print("Loaded ASTRA:", env_astra.shape)

# ---------------------------
# Load galaxy catalog
# ---------------------------
df = pd.read_csv(CATALOG)
print(f"Loaded {len(df):,} LRG galaxies")

# ---------------------------
# Rescale galaxy coordinates
# ---------------------------
def rescale(arr, old_min, old_max, new_min=0, new_max=640):
    return (arr-old_min)/(old_max-old_min)*(new_max-new_min)+new_min

df["x_scaled"] = rescale(df["x"], df["x"].min(), df["x"].max())
df["y_scaled"] = rescale(df["y"], df["y"].min(), df["y"].max())
df["z_scaled"] = rescale(df["z"], df["z"].min(), df["z"].max())

# ---------------------------
# Map galaxies to voxel indices
# ---------------------------
ix = np.clip((df["x_scaled"] / dx).astype(int), 0, N-1)
iy = np.clip((df["y_scaled"] / dx).astype(int), 0, N-1)
iz = np.clip((df["z_scaled"] / dx).astype(int), 0, N-1)

df["TWEB"]  = env_tweb[ix,iy,iz]
df["ASTRA"] = env_astra[ix,iy,iz]

# ---------------------------
# Build confusion matrix
# ---------------------------
conf = pd.crosstab(df["TWEB"], df["ASTRA"], rownames=["T-Web"], colnames=["ASTRA"])

print("\nCONFUSION MATRIX (counts):")
print(conf)

conf_pct = conf / conf.to_numpy().sum() * 100
print("\nCONFUSION MATRIX (%)")
print(conf_pct.round(2))

# ---------------------------
# Compute overall agreement
# ---------------------------
diag_sum = np.trace(conf.values)
total = conf.values.sum()
overall_agreement = diag_sum / total * 100

print(f"\n✅ Overall Agreement = {overall_agreement:.2f}%")

# ---------------------------
# Plot heatmap
# ---------------------------
plt.figure(figsize=(7,5))
sns.heatmap(
    conf_pct, annot=True, fmt=".1f",
    cmap="Blues",           # << improved bright colormap
    cbar_kws={"label":"Percentage [%]"},
    xticklabels=labels,
    yticklabels=labels
)

plt.title(f"ASTRA vs T-Web\nOverall Agreement = {overall_agreement:.2f}%",
          fontsize=14, weight="bold")
plt.xlabel("ASTRA Classification")
plt.ylabel("T-Web Classification")

plt.tight_layout()
outfig = os.path.join(FIG_DIR, "astra_tweb_confusion_matrix.png")
plt.savefig(outfig, dpi=450, bbox_inches="tight")
plt.show()

print(f"\n💾 Saved → {outfig}")
```

```python
# =====================================================
#  ASTRA-like Delaunay Triangulation with Zoom Panels (DESI ELG slice)
# =====================================================
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.spatial import Delaunay
from mpl_toolkits.axes_grid1.inset_locator import inset_axes, mark_inset

# -----------------------------------------------------
# 1. Load ELG galaxies
# -----------------------------------------------------
gal = pd.read_csv("lrg_galaxies_xyz.csv")
print(f"✅ Loaded {len(gal):,} galaxies")

# Thin slice in comoving z (±50 Mpc around median)
zmin, zmax = gal["z"].median() - 50, gal["z"].median() + 50
slice_gal = gal[(gal["z"] > zmin) & (gal["z"] < zmax)]
print(f"✅ Selected {len(slice_gal):,} galaxies in z ∈ [{zmin:.1f}, {zmax:.1f}]")

# -----------------------------------------------------
# 2. Sampling and random catalogue
# -----------------------------------------------------
N_obj = min(len(slice_gal), 2500)
sample = slice_gal.sample(N_obj, random_state=42)
x_obj, y_obj = sample["x"].values, sample["y"].values
xmin, xmax, ymin, ymax = x_obj.min(), x_obj.max(), y_obj.min(), y_obj.max()

N_rand = N_obj
x_rand = np.random.uniform(xmin, xmax, N_rand)
y_rand = np.random.uniform(ymin, ymax, N_rand)

# Combine into one array
X = np.concatenate([x_obj, x_rand])
Y = np.concatenate([y_obj, y_rand])
points = np.vstack([X, Y]).T
is_obj = np.zeros(len(points), dtype=bool)
is_obj[:N_obj] = True

# -----------------------------------------------------
# 3. Delaunay triangulation
# -----------------------------------------------------
tri = Delaunay(points)
edges = set()
for s in tri.simplices:
    for i in range(3):
        a, b = s[i], s[(i + 1) % 3]
        if a > b:
            a, b = b, a
        edges.add((a, b))
edges = np.array(list(edges))

# Classify edges
oo_edges = []
or_edges = []
for a, b in edges:
    if is_obj[a] and is_obj[b]:
        oo_edges.append((a, b))
    else:
        or_edges.append((a, b))
oo_edges = np.array(oo_edges)
or_edges = np.array(or_edges)

# -----------------------------------------------------
# 4. Main plot
# -----------------------------------------------------
fig, ax = plt.subplots(figsize=(8, 8))
ax.set_aspect("equal")
ax.set_xlim(xmin, xmax)
ax.set_ylim(ymin, ymax)

# Dashed random connections
for a, b in or_edges:
    ax.plot([points[a, 0], points[b, 0]], [points[a, 1], points[b, 1]],
            ls="--", lw=0.5, color="royalblue", alpha=0.45)
# Solid object connections
for a, b in oo_edges:
    ax.plot([points[a, 0], points[b, 0]], [points[a, 1], points[b, 1]],
            ls="-", lw=0.7, color="k", alpha=0.6)

ax.scatter(x_rand, y_rand, s=4, color="deepskyblue", label="Random points")
ax.scatter(x_obj, y_obj, s=8, color="k", label="DESI LRG galaxies")

ax.set_xlabel("x [Mpc]", fontsize=11)
ax.set_ylabel("y [Mpc]", fontsize=11)
ax.set_title("ASTRA-like Delaunay Triangulation — DESI LRG Slice", fontsize=13, weight="bold")
ax.legend(markerscale=2, fontsize=9, frameon=True)

# -----------------------------------------------------
# 5. Choose HD & LD regions (by hand or automatically)
# -----------------------------------------------------
# High-density: cluster center (lots of galaxies)
hd_center = [np.median(x_obj) - 0.1*(xmax-xmin), np.median(y_obj)]
# Low-density: outskirts (void)
ld_center = [np.median(x_obj) + 0.25*(xmax-xmin), np.median(y_obj) + 0.2*(ymax-ymin)]
box_size = 0.1 * (xmax - xmin)

# Draw rectangles
rect_hd = plt.Rectangle((hd_center[0]-box_size/2, hd_center[1]-box_size/2),
                        box_size, box_size, ec="red", lw=1.3, fc="none")
rect_ld = plt.Rectangle((ld_center[0]-box_size/2, ld_center[1]-box_size/2),
                        box_size, box_size, ec="red", lw=1.3, fc="none")
ax.add_patch(rect_hd)
ax.add_patch(rect_ld)

# -----------------------------------------------------
# 6. Create zoom-in inset axes
# -----------------------------------------------------
def zoom_inset(ax_main, center, label):
    x0, y0 = center
    ax_inset = inset_axes(ax_main, width="35%", height="35%",
                          loc="upper right" if label == "LD Region" else "upper left",
                          borderpad=1)
    ax_inset.set_xlim(x0 - box_size/2, x0 + box_size/2)
    ax_inset.set_ylim(y0 - box_size/2, y0 + box_size/2)
    ax_inset.set_aspect("equal")

    # Repeat the edge drawing inside the inset
    for a, b in or_edges:
        ax_inset.plot([points[a, 0], points[b, 0]], [points[a, 1], points[b, 1]],
                      ls="--", lw=0.5, color="royalblue", alpha=0.45)
    for a, b in oo_edges:
        ax_inset.plot([points[a, 0], points[b, 0]], [points[a, 1], points[b, 1]],
                      ls="-", lw=0.7, color="k", alpha=0.6)
    ax_inset.scatter(x_rand, y_rand, s=4, color="deepskyblue")
    ax_inset.scatter(x_obj, y_obj, s=8, color="k")
    ax_inset.set_xticks([]); ax_inset.set_yticks([])
    ax_inset.set_title(label, fontsize=9, weight="bold", color="red")

    # Connect inset to main rectangle
    mark_inset(ax_main, ax_inset,
               loc1=2 if label == "HD Region" else 1,
               loc2=4 if label == "HD Region" else 3,
               fc="none", ec="red", lw=0.8)

zoom_inset(ax, hd_center, "HD Region")
zoom_inset(ax, ld_center, "LD Region")

plt.tight_layout()
plt.savefig("output/astra_delaunay_desi_with_insets.png", dpi=450, bbox_inches="tight")
plt.show()

print("✅ Saved → output/astra_delaunay_desi_with_insets.png")
```
